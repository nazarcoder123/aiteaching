# what is an API ?

- An API stands for Application Programming Interface: It is an mechanism that allows two different software applications to communicate with each other and exchange data.

eg: Imagine you visit a hotel

1. customer looks at the menu.

2. Then place an order with waiter.

3. Then waiter communicate ur order with kitchen

4. The chef prepares the food

5. Then the waiter brings to you ur Food

- From the above waiter is acting as a API 
- Customer is acting as frontend 
- kitchen as Server(Backend) 
- Food Order as API Request 
- Food Prepare as API Response

# why do we need APIs ?

- To make real-world application
- To make UPI payment
- To check ur balance
- Food delivery App

# How does an API work?

website . Mobile App -> HTTP Request Get/users -> API/Backend (fetch-DB) -> Client recevies the data

- Client sends a request
- API recevies the request
- Backend processess the request
- Backend may communicate with a database
- API sends a response
- Client displays the response

# what is HTTP?

- HTTP stands for Hyper text transfer protocol.

Ex: http://example.com/users

# HTTP Methods

This is the most important concept should need to understand before FASTAPI.

- GET: purpose: Retrieve data from database eg: get employees name.
- POST: purpose: create data eg: Add a new student.
- PUT: purpose: Replace existing data eg: updating students details.
- PATCH: purpose: Partially update data eg: update students email. 
- DELETE: Purpose: delete data eg: delete a student details.

# what is an HTTP Status Code?

- 200:  okay

- 201: Created

- 400: Bad Request

- 401: Unauthorized

- 403: Forbidden

- 404: Not Found 

- 500: Internal Server Error

# what is REST API?

- Rest stands for representational state Transfer.

- Rest is an architectural style for designing network-based APIs
- Common Characteristics:

- Uses HTTP methods
- Uses resource such as students,employees and products
- commonly exchanges JSON data.
- Is generally stateless: each request carries the Infm needed to process it.

For Ex:

- GET:  /products
- POST: /products
- GET: /products/10
- DELETE: /products/10
