### What is a Python Set?
A **Set** is an unordered, mutable collection of unique elements of various data types.

### Key Characteristics:
1. **Unordered & Unindexed:** Elements have no fixed position; indexing (`s[0]`) is not supported.
2. **Unique Elements (Automatic Deduplication):** Duplicate values entered into a set are discarded automatically.
3. **Internal Hash Table Implementation:** Lookups, additions, and deletions execute in $O(1)$ average time complexity.
4. **Element Constraint:** Elements inside a set must be **hashable and immutable** (numbers, strings, tuples). Unhashable types like lists and dictionaries cannot be stored inside sets.
5. **Can Store `None`:** Sets can hold the `None` value alongside other elements.

---
## 1. Creating a Set

Sets can be created using:
- Curly braces `{...}` with comma-separated values.
- The built-in [`set()`](https://www.geeksforgeeks.org/python/python-set-function/) constructor.

> **Crucial Rule for Empty Sets:**  
> To create an empty set, you **must use `set()`**. Writing `{}` creates an **empty dictionary**, not a set!

```python
# 1. Basic Set using curly braces {}
s = {1, 2, 3, 4}
print("Set s:", s)

# 2. Empty Set vs Empty Dictionary
empty_dict = {}
empty_set = set()

print("\nType of {}    :", type(empty_dict))
print("Type of set() :", type(empty_set))
```

---
## 2. Typecasting Objects into Sets with `set()`

The `set()` constructor can convert any iterable (strings, lists, tuples, dictionaries) into a set:
- **From String:** Splits into unique individual characters.
- **From List/Tuple:** Automatically removes all duplicate entries.
- **From Dictionary:** Extracts only the unique **keys** of the dictionary.

```python
# Typecasting different iterables into sets using set()

# Empty set
s_empty = set()
print("Empty set:", s_empty)

# From String (splits into unique characters)
s_str = set("GeeksForGeeks")
print("From String 'GeeksForGeeks':", s_str)

# From List (deduplicates items)
s_list = set(["GFG", "For", "Geeks"])
print("From List:", s_list)

# From Tuple
t = ("GFG", "for", "Geeks")
print("From Tuple:", set(t))

# From Dictionary (extracts keys only)
d = {"GFG": 1, "for": 2, "Geeks": 3}
print("From Dictionary (Keys only):", set(d))
```

---
## 3. Unordered, Unindexed Nature & Immutability of Elements

1. **No Guaranteed Order:** The order in which elements appear in output may vary across runs.
2. **Automatic Deduplication:** Duplicate items are discarded immediately.
3. **No Indexing Support:** Trying to access `s[0]` raises a [`TypeError: 'set' object is not subscriptable`](https://www.geeksforgeeks.org/python/handling-typeerror-exception-in-python/).

```python
# Demonstrating Deduplication and Indexing Restriction
s = {3, 1, 4, 1, 5, 9, 2}

print("Deduplicated Set:", s)

# Attempting indexing on a set
try:
    print(s[0])
except TypeError as err:
    print("\n[Caught Expected Error]:", err)
    print("Explanation: Sets are unordered and do not support index lookups.")
```

## 4. Adding Elements to a Set

- **[`add(element)`](https://www.geeksforgeeks.org/python/set-add-python/):** Adds a **single** element to the set.
- **[`update(iterable)`](https://www.geeksforgeeks.org/python/python-set-update/):** Adds **multiple** elements from an iterable (list, tuple, or another set).

```python
# Adding elements: add() vs update()
s = {1, 2, 3}
print("Initial set:", s)

# Adding a single item with add()
s.add(4)
print("After s.add(4):", s)

# Adding multiple items with update()
s.update([5, 6])
print("After s.update([5, 6]):", s)

# Adding another set via update()
s.update({7, 8})
print("After s.update({7, 8}):", s)
```

---
## 5. Accessing Set Elements & Membership Testing

Because sets are unindexed, elements are accessed by:
1. Iterating through the set using a `for` loop.
2. Using the **`in`** and **`not in`** operators for $O(1)$ fast membership testing.

```python
# Iterating and Membership Testing
s = {"Geeks", "For", "Geeks"}

print("Iterating through set elements:")
for item in s:
    print("-", item)

print("\nMembership checks ($O(1)$ lookup time):")
print("'Geeks' in s     :", "Geeks" in s)
print("'Python' in s    :", "Python" in s)
print("'Python' not in s:", "Python" not in s)
```

---
## 6. Removing Elements from a Set

Python offers four methods to remove elements:

| Method | Behavior when element is missing |
| :--- | :--- |
| **`remove(x)`** | Removes `x`. **Raises `KeyError`** if `x` is not in the set. |
| **`discard(x)`** | Removes `x`. **Does NOT raise an error** if `x` is not in the set. |
| **`pop()`** | Removes and returns an **arbitrary element**. Raises `KeyError` if set is empty. |
| **`clear()`** | Removes **all** elements, leaving an empty set `set()`. |

```python
# Demonstrating remove(), discard(), pop(), and clear()
s = {1, 2, 3, 4, 5}

# 1. remove()
s.remove(3)
print("After remove(3):", s)

# remove() raises KeyError when element is missing:
try:
    s.remove(10)
except KeyError as err:
    print("Caught KeyError with remove(10):", err)

# 2. discard() safely ignores missing elements
s.discard(4)
print("\nAfter discard(4) :", s)
s.discard(10)  # No error raised!
print("After discard(10):", s)

# 3. pop() removes an arbitrary element
popped_val = s.pop()
print(f"\nAfter pop()      : {s} (Popped element: {popped_val})")

# 4. clear() empties the set
s.clear()
print("After clear()    :", s)

# pop() on empty set raises KeyError
try:
    s.pop()
except KeyError as err:
    print("Caught KeyError on empty pop():", err)
```

## 7. Frozen Sets (`frozenset`)

A [`frozenset`](https://www.geeksforgeeks.org/python/frozenset-in-python/) is an **immutable** version of a Python set.
- Once created, elements **cannot** be added, removed, or modified.
- Like sets, frozensets automatically discard duplicate elements.
- **Key Advantage:** Because they are immutable and hashable, `frozenset` objects can be used as **keys in dictionaries** or as elements in other sets.

```python
# Demonstrating frozenset
fs1 = frozenset([1, 2, 3, 4, 5])
print("fs1:", fs1)

s = {3, 1, 4, 1, 5}
fs2 = frozenset(s)
print("fs2 (from set):", fs2)

# Attempting mutation on a frozenset raises AttributeError:
try:
    fs1.add(6)
except AttributeError as err:
    print("\n[Caught Expected Error]:", err)
    print("frozenset objects are immutable and have no .add() or .remove() methods.")

# Using frozenset as a dictionary key (valid because it is hashable):
group_roles = {
    frozenset(["Alice", "Bob"]): "Admins",
    frozenset(["Charlie"]): "Guest"
}
print("\nFrozenset as Dict Key:", group_roles)
```

## 8. Mathematical Set Operations

Sets support powerful algebraic operations:
- **Union (`|`):** Elements present in either set $A \cup B$.
- **Intersection (`&`):** Elements present in both sets $A \cap B$.
- **Difference (`-`):** Elements in $A$ but not in $B$ ($A \setminus B$).
- **Symmetric Difference (`^`):** Elements in either set, but not in both.

```python
# Demonstrating Mathematical Set Operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Set A:", set_a)
print("Set B:", set_b)
print("-" * 40)

print("Union (set_a | set_b)               :", set_a | set_b)
print("Intersection (set_a & set_b)        :", set_a & set_b)
print("Difference (set_a - set_b)          :", set_a - set_b)
print("Symmetric Difference (set_a ^ set_b):", set_a ^ set_b)
```
