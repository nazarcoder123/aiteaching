### What are Conditional Statements?
**Conditional statements** are control flow structures used to govern the path of program execution based on specific conditions.

They evaluate expressions to a boolean result (`True` or `False`), allowing programs to make decisions dynamically and execute different code blocks accordingly.

### Core Control Structures in Python:
1. `if` statement
2. Short Hand `if`
3. `if-else` statement
4. `if-elif-else` ladder
5. Nested `if-else` statements
6. Conditional Expression (Ternary Operator)
7. Structural Pattern Matching (`match-case`)

---
## 1. The `if` Statement

The `if` statement executes an indented block of code **only if** the given condition evaluates to `True`. If the condition evaluates to `False`, the code block is bypassed.

### Syntax:
```python
if condition:
    # indented code block to execute if True
```

````python
# Demonstration of the basic 'if' statement
age = 20

if age >= 18:
    print("Eligible to vote.")

# Condition that evaluates to False (skipped):
if age < 18:
    print("This line will not execute.")
````

---
## 2. Short Hand `if` (One-Line `if`)

When only a **single statement** needs to be executed within the `if` block, Python allows placing it on the same line after the colon.

### Syntax:
```python
if condition: statement
```

````python
# Demonstration of Short Hand 'if'
age = 19

if age > 18: print("Eligible to Vote.")
````

---
## 3. The `if-else` Statement

The [`if-else`](https://www.geeksforgeeks.org/python/python-if-else/) statement provides a two-way branch:
- Executes the `if` block when the condition is `True`.
- Executes the `else` block when the condition is `False`.

### Syntax:
```python
if condition:
    # block executed if True
else:
    # block executed if False
```

````python
# Demonstration of 'if-else' statement
age = 10

if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
````

---
## 4. The `if-elif-else` Ladder

When you need to test **multiple mutually exclusive conditions**, the `elif` (short for *else if*) statement is used.
- Python evaluates conditions sequentially from top to bottom.
- The **first condition** that evaluates to `True` runs its code block.
- All remaining subsequent conditions are skipped.
- If none of the conditions evaluate to `True`, the optional `else` block executes.

````python
# Demonstration of 'if-elif-else' ladder
age = 25

if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")

# Explanation:
# 1. age <= 12 (25 <= 12) -> False (skipped)
# 2. age <= 19 (25 <= 19) -> False (skipped)
# 3. age <= 35 (25 <= 35) -> True  (executes "Young adult.")
# 4. The subsequent 'else' block is automatically skipped.
````

---
## 5. Nested `if-else` Statements

A [Nested `if-else`](https://www.geeksforgeeks.org/python/nested-if-statement-in-python/) is an `if` or `else` block placed inside another `if` or `else` block. It is used when a secondary decision depends on the outcome of a primary condition.

### Combining Conditions with Logical Operators
Instead of deep nesting, multiple conditions can often be flattened using `and`, `or`, and `not`.
*For detailed patterns, see [Check Multiple conditions in if statement](https://www.geeksforgeeks.org/python/check-multiple-conditions-in-if-statement-python/).*

````python
# Demonstration of Nested 'if-else'
age = 70
is_member = True

if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")

# Equivalent logic using compound conditions (Logical 'and'):
print("-" * 40)
if age >= 60 and is_member:
    print("Combined check: 30% senior discount applied!")
````

---
## 6. Conditional Expression (Ternary Operator)

The [Ternary Operator](https://www.geeksforgeeks.org/python/ternary-operator-in-python/) provides a concise syntax for assigning values or returning results based on a condition in a single readable line.

### Syntax:
```python
value_if_true if condition else value_if_false
```

````python
# Demonstration of Conditional Expression (Ternary Operator)
age = 20

# Assigning status based on age
status = "Adult" if age >= 18 else "Minor"
print("Status:", status)

# Direct inline evaluation inside print():
score = 82
print("Result:", "Passed" if score >= 50 else "Failed")
````

---
## 7. Match-Case Statement (Structural Pattern Matching)

Introduced in **Python 3.10+**, the [`match-case` statement](https://www.geeksforgeeks.org/python/python-match-case-statement/) compares a value against multiple patterns and executes the corresponding block.

It serves as Python's modern, powerful equivalent to the `switch-case` construct found in other languages.

### Key Syntax Features:
- `case pattern:`: Matches a specific value or expression.
- `case pattern1 | pattern2:`: The pipe `|` operator matches **either** pattern.
- `case _:`: The underscore `_` acts as a **wildcard / default case** when no other pattern matches.

````python
# Demonstration of Python 3.10+ Match-Case statement
number = 2

match number:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("Other number")

# Another example: HTTP Status Code matcher
http_code = 404

match http_code:
    case 200:
        print("Status: 200 OK")
    case 400 | 404:
        print(f"Status: Client Error ({http_code})")
    case 500:
        print("Status: Server Error")
    case _:
        print("Status: Unknown Code")
````

- Match `"Saturday"` or `"Sunday"` and print `"Weekend!"`
- Match `"Monday"` through `"Friday"` and print `"Weekday"`
- Match any other input with `case _` and print `"Invalid day"`

````python
# --- Student Practice Solutions ---

# Exercise 1: Traffic Light
light_color = "Yellow"
if light_color == "Red":
    print("Stop")
elif light_color == "Yellow":
    print("Caution / Prepare to stop")
elif light_color == "Green":
    print("Go")
else:
    print("Invalid Signal")

print("-" * 45)

# Exercise 2: Grade Calculator
marks = 78
if marks >= 90:
    grade = "Grade A"
elif marks >= 80:
    grade = "Grade B"
elif marks >= 70:
    grade = "Grade C"
else:
    grade = "Needs Improvement"
print(f"Marks: {marks} -> Result: {grade}")

print("-" * 45)

# Exercise 3: Match-Case Day Classifier
day = "Saturday"
match day:
    case "Saturday" | "Sunday":
        print(f"{day} is a Weekend!")
    case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday":
        print(f"{day} is a Weekday.")
    case _:
        print("Invalid day entered.")
````
