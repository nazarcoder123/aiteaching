### What is a Variable?
Variables are used to **store data** that can be referenced and manipulated during program execution. A variable is essentially a **name assigned to a value**.

Unlike Java, C++, and many statically-typed languages, **Python variables do not require explicit declaration of data type**. The type of the variable is **automatically inferred** based on the value assigned to it.

````python
# Basic Variable Declaration and Printing
x = 5
name = "Alex"

print(x)
print(name)

# In Python, types are inferred automatically:
print("Type of x:", type(x))
print("Type of name:", type(name))
````

---
## Rules for Naming Variables

To use variables correctly in Python, the following naming rules must be followed:

1. **Allowed Characters:** Names can contain letters (`a-z`, `A-Z`), digits (`0-9`), and underscores (`_`).
2. **First Character:** The first character **cannot be a digit**. It must begin with a letter or an underscore.
3. **Case-Sensitivity:** Names are case-sensitive (`myVar` and `myvar` are completely distinct variables).
4. **Reserved Keywords:** [Keywords](https://www.geeksforgeeks.org/python/python-keywords/) such as `if`, `else`, `for`, `class`, `def`, `while`, etc., **cannot** be used as variable names.

| Valid Variable Names | Reason |
| :--- | :--- |
| `age = 21` | Standard letters |
| `_colour = "lilac"` | Can start with underscore |
| `total_score = 90` | Can combine words with underscores (snake_case) |

| Invalid Variable Names | Error Reason |
| :--- | :--- |
| `1name = "Error"` | Starts with a digit |
| `class = 10` | `class` is a reserved keyword |
| `user-name = "Doe"` | Hyphen `-` is interpreted as subtraction |

````python
# ✅ Valid Variable Names
age = 21
_colour = "lilac"
total_score = 90

print("Age:", age)
print("Colour:", _colour)
print("Total score:", total_score)
````

````python
# ❌ Invalid Variable Names
# Uncomment each line one by one to observe the SyntaxError!

# 1name = "Error"     # SyntaxError: invalid decimal literal (cannot start with a digit)
# class = 10         # SyntaxError: invalid syntax ('class' is a reserved keyword)
# user-name = "Doe"   # SyntaxError: cannot assign to expression (hyphen treated as minus)

# Let's inspect reserved keywords in Python:
import keyword

print("Total reserved keywords in Python:", len(keyword.kwlist))
print("Sample keywords:", keyword.kwlist[:8])
print("Is 'class' a keyword?", keyword.iskeyword("class"))
print("Is 'my_var' a keyword?", keyword.iskeyword("my_var"))
````

## Assigning Values to Variables

Python provides flexible ways to assign values to variables:

### 1. Basic Assignment
Variables are assigned values using the `=` [operator](https://www.geeksforgeeks.org/python/python-operators/).
```python
x = 5
y = 3.14
z = "Hi"
```

### 2. Dynamic Typing
Python is **dynamically typed**, which means the same variable can hold different data types during program execution without throwing an error.
```python
x = 10            # integer
x = "Now a string" # reassigned to string
```

### 3. Assigning Same Value to Multiple Variables (Chained Assignment)
The same value can be assigned to multiple variables simultaneously in a single line.
```python
a = b = c = 100
```

### 4. Assigning Different Values in a Single Line (Multiple Assignment)
Multiple variables can be assigned distinct values simultaneously using comma separation.
```python
x, y, z = 1, 2.5, "Python"
```

````python
# 1. Basic Assignment
x = 5
y = 3.14
z = "Hi"
print("Basic:", x, y, z)

# 2. Dynamic Typing Demonstration
x = 10
print("x as integer:", x, "| Type:", type(x))

x = "Now a string"
print("x as string: ", x, "| Type:", type(x))

# 3. Assigning the Same Value (Chained Assignment)
a = b = c = 100
print("Chained assignment:", a, b, c)

# 4. Assigning Different Values (Multiple Assignment)
x, y, z = 1, 2.5, "Python"
print("Multiple assignment:", x, y, z)
````

## Concept of Object Reference & Memory Model

In Python, variables **do not store values directly**. Instead, variables store **references (pointers) to objects in memory**.

### Step 1: Initial Assignment
```python
x = 5
```
When `x = 5` is executed:
- Python creates an integer object `5` in heap memory.
- `x` is bound to reference this object.

### Step 2: Shared Reference
```python
y = x
```
This statement creates `y` and makes it reference the **same object** as `x`, not `x` itself.
This is called a **[Shared Reference](https://www.geeksforgeeks.org/python/shared-reference-in-python/)**, where multiple variables point to the identical object in memory.

### Step 3: Reassigning `x`
```python
x = 'Geeks'
```
Python creates a new object `'Geeks'` and makes `x` reference it.
The variable `y` **remains unchanged**, still referencing the original integer object `5`.

### Step 4: Reassigning `y` & Garbage Collection
```python
y = "Computer"
```
A new object `'Computer'` is assigned to `y`. The previous integer object `5` now has zero active references and becomes eligible for **automatic Garbage Collection (GC)**.

> **Key Takeaway:** When a variable is reassigned, it starts referencing a new object while the old unreferenced object is reclaimed by Python's garbage collector.

---
### Variable Scope
The **scope** of a variable determines where in the program it can be accessed (e.g., inside a function vs. across the entire script).
*For detailed information on scope, refer to [Python Scope of Variables](https://www.geeksforgeeks.org/python/python-scope-of-variables/).*

````python
# Demonstrating Object References using id() (Memory Address)

# Step 1: Assign x = 5
x = 5

# Step 2: Shared Reference: y = x
y = x

print("--- Step 1 & 2: Shared Reference ---")
print("x value:", x, "| Memory id(x):", id(x))
print("y value:", y, "| Memory id(y):", id(y))
print("Do x and y share the exact same object in memory?", x is y)

# Step 3: Reassigning x = 'Geeks'
x = 'Geeks'
print("\n--- Step 3: Reassign x ---")
print("x value:", x, "| Memory id(x):", id(x))
print("y value:", y, "| Memory id(y):", id(y), "(y still references 5!)")

# Step 4: Reassigning y = 'Computer'
y = "Computer"
print("\n--- Step 4: Reassign y ---")
print("y value:", y, "| Memory id(y):", id(y))
print("Original object 5 is now dereferenced and eligible for Garbage Collection.")
````

## Understanding Variable Reassignment

In this example, we verify whether modifying one variable affects another when both initially reference the same object:

```python
x = 1
y = x
y = y + 1

print(x)
print(y)
```

### Explanation:
1. Initially, both `x` and `y` reference the object `1`.
2. When `y = y + 1` runs, Python computes `1 + 1 = 2`, allocates a new integer object `2`, and binds `y` to it.
3. `x` continues to reference the integer `1`. Changing `y` does **not** affect `x`!

````python
# Variable Reassignment Trace
x = 1
y = x
y = y + 1

print("Value of x:", x)
print("Value of y:", y)

# Let's inspect the memory IDs:
print("Memory ID of x:", id(x))
print("Memory ID of y:", id(y))
````

## Deleting a Variable (`del` Keyword)

The [`del`](https://www.geeksforgeeks.org/python/python-del-to-delete-objects/) keyword is used to remove a variable name binding from the current namespace. After deletion, the variable can **no longer be accessed**.

If you attempt to access a deleted variable, Python raises a `NameError`:
```
ERROR! Traceback (most recent call last):
  File "<main.py>", line 3, in <module>
NameError: name 'x' is not defined
```

````python
# Deleting a Variable using del
x = 10
print("Value before deletion:", x)

# Delete variable x from memory
del x

# Accessing x after deletion raises NameError:
try:
    print(x)
except NameError as error:
    print("\n[Expected Error Caught]:")
    print("NameError:", error)
    print("Explanation: 'x' was deleted and no longer exists in scope!")
````

## Practical Examples

### 1. Swapping Two Variables
In traditional programming languages, swapping values requires a temporary third variable (`temp`).  
In Python, you can swap values directly in a single clean line using **multiple assignment / tuple unpacking**:
```python
a, b = b, a
```

### 2. Counting Characters in a String
Assigning the result of built-in functions (like `len()`) to variables allows you to store and reuse calculated values:
```python
word = "Python"
length = len(word)
```

````python
# Practical Example 1: Swapping two variables without a temp variable
a, b = 5, 10
print(f"Before swap: a = {a}, b = {b}")

# Pythonic Swap:
a, b = b, a
print(f"After swap:  a = {a}, b = {b}")

print("-" * 40)

# Practical Example 2: Counting characters in a string
word = "Python"
length = len(word)

print("Word:", word)
print("Length of the word:", length)
````
