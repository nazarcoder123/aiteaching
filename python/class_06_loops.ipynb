### What are Loops?
In programming, **Loops** are control flow structures used to execute a block of code repeatedly until a specified condition is satisfied or all elements of a collection have been processed.

Loops eliminate repetitive code, streamline data processing, and implement iterative algorithms efficiently.

### Primary Types of Loops in Python:
1. **`for` Loop:** Definite iteration used to traverse sequences (lists, tuples, strings, ranges).
2. **`while` Loop:** Indefinite/conditional iteration that continues as long as a test condition remains `True`.

---
## 1. The `for` Loop

A [For loop](https://www.geeksforgeeks.org/python/python-for-loops/) is used to iterate over a sequence (such as a string, list, tuple, or range). It executes its block of code once for each element in the sequence.

### Generating Number Ranges: `range(start, stop, step)`
- `range(stop)`: Generates numbers from `0` up to `stop - 1`.
- `range(start, stop)`: Generates numbers from `start` up to `stop - 1`.
- `range(start, stop, step)`: Increments by `step` on each iteration.

````python
# Basic For Loop with range()
n = 4

for i in range(0, n):
    print(i)

# Explanation:
# 1. n = 4 defines the upper bound (exclusive).
# 2. range(0, n) yields integers: 0, 1, 2, 3.
# 3. 'for i in ...' assigns each number to 'i' successively and prints it.
````

---
## 2. Iterating Through Sequences

There are two primary ways to iterate through elements of a collection:
1. **Iterating by Index:** Using [`len()`](https://www.geeksforgeeks.org/python/python-len-function/) and [`range()`](https://www.geeksforgeeks.org/python/python-range-function/).
2. **Direct Item Iteration (Pythonic):** Traversing items directly without index overhead.
3. **Using `enumerate()`:** Retrieving both the index and the item simultaneously.

````python
# Method 1: Iterating by Index using range(len(...))
a = ["geeks", "for", "geeks"]

print("--- Iterating by Index ---")
for idx in range(len(a)):
    print(f"Index {idx}: {a[idx]}")

print("\n--- Method 2: Direct Item Iteration (Pythonic) ---")
for item in a:
    print(item)

print("\n--- Method 3: Using enumerate() for Index & Item ---")
for idx, val in enumerate(a):
    print(f"Position {idx} -> {val}")
````

## 3. The `while` Loop

A [While loop](https://www.geeksforgeeks.org/python/python-while-loop/) repeatedly executes a block of code **as long as the given condition remains `True`**.

As soon as the condition evaluates to `False`, the loop terminates and execution proceeds to the line immediately following the loop.

### Structure of a While Loop:
1. **Initialization:** Set up the counter variable (e.g., `cnt = 0`).
2. **Condition Test:** Check if the loop should continue (e.g., `cnt < 3`).
3. **Update / Increment:** Modify the counter to eventually make the condition `False` (e.g., `cnt += 1`).

````python
# Demonstration of While Loop
cnt = 0

while (cnt < 3):
    cnt = cnt + 1
    print("Hello Geek")

print(f"Loop finished! Final counter value: {cnt}")
````

---
## 4. Infinite While Loops & Safe Termination

An **infinite loop** runs indefinitely because its condition never becomes `False`.

```python
# WARNING: An infinite loop runs forever!
while True:
    print("Hello Geek")
```

> **Important Note:** Avoid unintentional infinite loops as they consume system resources and freeze execution.
> When an intentional `while True:` loop is needed (such as in server event loops or user menus), always provide a **conditional exit using the `break` statement**.

````python
# Controlled Infinite Loop using 'break' to terminate safely
step = 0

while True:
    step += 1
    print(f"Step {step}: Running safely...")
    
    # Safe exit condition
    if step >= 3:
        print("Termination condition reached. Breaking out of loop.")
        break

print("Safely exited the while True loop!")
````

---
## 5. Loop Control Statements: `break`, `continue`, `pass`, and `else`

Python provides special statements to alter loop behavior:

| Statement | Purpose |
| :---: | :--- |
| **`break`** | Terminates the loop immediately and jumps to the code outside. |
| **`continue`** | Skips the remainder of the current iteration and jumps directly to the next iteration. |
| **`pass`** | Acts as a syntactic placeholder (does nothing). |
| **`else` with loops** | Executes **only if** the loop completed all iterations naturally (without being interrupted by `break`). |

````python
# Demonstration of break, continue, and loop-else

print("--- Demonstration of 'continue' ---")
for num in range(1, 6):
    if num == 3:
        print("Skipping 3 via continue!")
        continue
    print("Number:", num)

print("\n--- Demonstration of 'break' & loop 'else' ---")
search_target = 7
numbers = [1, 3, 5, 7, 9]

for val in numbers:
    if val == search_target:
        print(f"Found target {search_target}! Breaking out.")
        break
else:
    print("Target not found in sequence.")
````

## 6. Nested Loops

A [Nested loop](https://www.geeksforgeeks.org/python/python-nested-loops/) is a loop placed inside another loop.
- The **outer loop** advances once per cycle.
- For each outer loop cycle, the **inner loop** executes through all of its iterations from start to finish.

````python
# Nested Loops: Pattern Printing
for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()  # newline after completing inner loop

# Explanation:
# i=1 -> inner loop runs 1 time  -> prints: 1
# i=2 -> inner loop runs 2 times -> prints: 2 2
# i=3 -> inner loop runs 3 times -> prints: 3 3 3
# i=4 -> inner loop runs 4 times -> prints: 4 4 4 4
````

````python
# Nested Loops: Iterating a 2D Grid / Matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Iterating through 2D Grid:")
for row_idx, row in enumerate(matrix):
    for col_idx, val in enumerate(row):
        print(f"matrix[{row_idx}][{col_idx}] = {val}", end=" | ")
    print()
````

### Exercise 1: Sum of First N Numbers
Write a `while` loop to calculate the sum of numbers from `1` to `10` ($1 + 2 + 3 + ... + 10$).

### Exercise 2: Multiplication Table with For Loop
Use a `for` loop to generate and print the multiplication table for number `5` (from $5 \times 1$ to $5 \times 10$).

### Exercise 3: Inverted Star Pyramid (Nested Loop)
Write nested `for` loops to print an inverted right-angled triangle pattern of stars (`*`):
```
* * * *
* * *
* *
*
```

````python
# --- Student Practice Solutions ---

# Exercise 1: Sum using while loop
n = 10
total_sum = 0
current = 1

while current <= n:
    total_sum += current
    current += 1

print(f"Sum of first {n} numbers: {total_sum}")

print("-" * 40)

# Exercise 2: Multiplication table of 5
multiplier = 5
print(f"Multiplication Table for {multiplier}:")
for i in range(1, 11):
    print(f"{multiplier} x {i:>2} = {multiplier * i:>2}")

print("-" * 40)

# Exercise 3: Inverted Star Pattern
rows = 4
print("Inverted Star Pattern:")
for i in range(rows, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
````
