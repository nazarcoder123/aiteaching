```python
import os
import requests

def get_weather_owm(city_name: str) -> dict:
    """
    Fetches real-time weather details for a specified city.

    Args:
        city_name (str): The name of the city to query (e.g., "Hyderabad" or "London").

    Returns:
        dict: A dictionary containing weather details with the following keys on success:
            - "city" (str): Resolved city name.
            - "country" (str): Two-letter country code.
            - "temperature" (str): Current temperature in Celsius (°C).
            - "feels_like" (str): Human-perceived temperature in Celsius (°C).
            - "humidity" (str): Relative humidity percentage.
            - "pressure" (str): Atmospheric pressure in hPa.
            - "wind_speed" (str): Wind speed in meters per second (m/s).
            - "description" (str): Short text description of the weather condition.
            
            If the request fails, returns a dictionary with a single key:
            - "error" (str): Description of the error that occurred.

    Example:
        >>> weather = get_weather_owm("Hyderabad")
        >>> print(weather.get("temperature"))
        '28.5°C'
    """
    # Replace "YOUR_OPENWEATHERMAP_API_KEY" with your actual key, 
    # or set it in your environment variables as OWM_API_KEY
    api_key = os.getenv("OWM_API_KEY", "")
    
    base_url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        "q": city_name.strip(),
        "appid": api_key,
        "units": "metric"
    }

    try:
        response = requests.get(base_url, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()
        
        return {
            "city": data["name"],
            "country": data["sys"]["country"],
            "temperature": f"{data['main']['temp']}°C",
            "feels_like": f"{data['main']['feels_like']}°C",
            "humidity": f"{data['main']['humidity']}%",
            "pressure": f"{data['main']['pressure']} hPa",
            "wind_speed": f"{data['wind']['speed']} m/s",
            "description": data["weather"][0]["description"].title()
        }
    except requests.exceptions.HTTPError as http_err:
        if response.status_code == 404:
            return {"error": f"City '{city_name}' not found. Please check the spelling."}
        elif response.status_code == 401:
            return {"error": "Invalid API key. Please check your OpenWeatherMap API key."}
        return {"error": f"HTTP error occurred: {http_err}"}
    except requests.exceptions.RequestException as err:
        return {"error": f"Network error: {err}"}


# Example usage (User enters city name via input or direct call):
if __name__ == "__main__":
    city = input("Enter city name: ")
    weather_details = get_weather_owm(city)
    
    for key, value in weather_details.items():
        print(f"{key.replace('_', ' ').title()}: {value}")
```

# Py Execution Models: Sync , Async , Multi-Threading & parallel processing 

> Multi-threading: running multiple threads within the same process

> shared memory: Threads share the same memory and resource of their parent process

> Multiple Workers: 

Real Life eg:

Imagine ur working in a restaurant:

> Thread 1: Takes an order

> Thread 2: Prepare a bill

> Thread 3: Ans the call

while one task is waiting for another can make progress

```python
import threading 
import time 

def task(name): 
    print(f"{name} started")
    time.sleep(2) # waiting for response 
    print(f"{name} completed") 

start = time.time() 

t1 = threading.Thread(target=task, args=("Task A",))
t2 = threading.Thread(target=task, args=("Task B",))
t3 = threading.Thread(target=task, args=("Task C",))


t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

print(f"Total time: {time.time() - start:.2f} seconds")
```

# Parallel Processing

Parallel processing means executing multiple tasks simultaneously using multiple processes often on different CPU

each have there own seprate memory space

> Example:

> worker 1: Calc numbers in Task A.

> worker 2: Calc numbers in Task B.

> worker 3: Calc numbers in Task C.

> worker 4: Calc numbers in Task D.

All four workers can perform CPU-intensive calculations at the same time provided sufficient CPU cores and resources are available.



```python
import multiprocessing
import os
import time

def addition(a, b):
    print(f"Addition running | PID: {os.getpid()}")
    result = a + b
    print(f"Addition result: {result}")

def subtraction(a, b):
    print(f"Subtraction running | PID: {os.getpid()}")
    result = a - b
    print(f"Subtraction result: {result}")

def multiplication(a, b):
    print(f"Multiplication running | PID: {os.getpid()}")
    result = a * b
    print(f"Multiplication result: {result}")

if __name__ == "__main__":
    # Get inputs (demonstrated with sample values for notebook)
    num1 = 10.0
    num2 = 5.0
    print(f"Inputs: num1 = {num1}, num2 = {num2}\n")

    start = time.time()

    # Create separate processes
    p1 = multiprocessing.Process(
        target=addition, args=(num1, num2)
    )

    p2 = multiprocessing.Process(
        target=subtraction, args=(num1, num2)
    )

    p3 = multiprocessing.Process(
        target=multiplication, args=(num1, num2)
    )

    # Start all processes
    p1.start()
    p2.start()
    p3.start()

    # Wait for all processes to finish
    p1.join()
    p2.join()
    p3.join()

    print(f"\nTotal execution time: {time.time() - start:.2f} seconds")
```

# Sync

Synchronous(Sequential): One waiter, one order at a time (sequential blocking on a single thread)

Eg: Imagine a hotel has one waiter

1. The waiter takes the tea order.
2. The waiter waits until the tea is prepared and served.
3. Only then does the waiter take the biryani order.
4. After that the waiter handles the coffee order


```python
import time
import threading

def make_tea():
    print(f"Making tea...     | Thread: {threading.current_thread().name}")
    time.sleep(2)  # Wait for 2 seconds (Blocking)
    print("Tea is ready!")

def make_coffee():
    print(f"Making coffee...  | Thread: {threading.current_thread().name}")
    time.sleep(1)  # Wait for 1 second (Blocking)
    print("Coffee is ready!")

start = time.time()

# Execute tasks one by one (Synchronously)
make_tea()
make_coffee()

print(f"All tasks completed in: {time.time() - start:.2f} seconds")
```

# Async

Asychronous: when we are having multiple function we provided the I/N rather than waiting for another function O/P it will return its result independently

Example: The smart async waiter

Imagine the same waiter but this time the waiter works like this:

1. sends the tea order to the kitchen
2. Instead of waiting Idly , Immediately takes the biryani order
3. sends the biryani order to the kitchen amd Immeditaly takes the coffee order
4. comes back to the serve each order when it is notified that the dish is ready

```python
import asyncio
import time
import threading

async def order(food):
    print(f"Ordered: {food:<7} | Thread: {threading.current_thread().name}")
    await asyncio.sleep(3)  # Non-blocking simulated wait for kitchen
    print(f"Ready:   {food:<7} | Thread: {threading.current_thread().name}")

async def main():
    # Run all three orders concurrently inside the Single-Threaded Event Loop
    await asyncio.gather(
        order("Tea"),
        order("Biryani"),
        order("Coffee")
    )

start = time.time()
asyncio.run(main())
print(f"Total time: {time.time() - start:.2f} seconds (Single Thread Concurrency!)")
```
