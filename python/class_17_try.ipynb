# Python Try Except

- Errors and exceptions can be interrupt the normal flow of a programe during exection. python provides try and except blocks to handle such errors gracefully without stopping the entire programe.

```python
# Before try-except - what happens when an errors occurs ?

print("programe started") 

a = 10
b = 0 

result = a / b 
print("Result:", result) 
print("programe completed") 
```

```python
# when we use try-except 

print("Programe started") 

try:
    a = 10
    b = 0 
    result = a / b 

    print("Result:", result) 
except:
    print("somethings went wrong.") 
print("programe completed") 
```

```python
# without try-except 

age = int(input("Enter your age: ")) 
print("Your age is:",age) 
```

```python
# with try-except 

try:
    age = int(input("Enter your age: ")) 
    print("Your age is", age) 
except:
    print("please enter a valid number!") 
print("Thanks you!")
```

```python
# Before try-except
import math

def perform_all_operations(a: int, b: int) -> dict:
    """
    Performs addition, subtraction, multiplication, division, 
    and square root on two integer inputs.
    """
    # 1. Addition
    addition = a + b

    # 2. Subtraction
    subtraction = a - b

    # 3. Multiplication
    multiplication = a * b

    # 4. Division (safe check for zero)
    division = a / b if b != 0 else "Undefined (cannot divide by zero)"

    # 5. Square Roots (safe check for negative values)
    sqrt_a = math.isqrt(a) if a >= 0 and math.isqrt(a)**2 == a else (round(math.sqrt(a), 4) if a >= 0 else "Undefined (negative number)")
    sqrt_b = math.isqrt(b) if b >= 0 and math.isqrt(b)**2 == b else (round(math.sqrt(b), 4) if b >= 0 else "Undefined (negative number)")

    return {
        "Addition (a + b)": addition,
        "Subtraction (a - b)": subtraction,
        "Multiplication (a * b)": multiplication,
        "Division (a / b)": division,
        "Square Root of a": sqrt_a,
        "Square Root of b": sqrt_b,
    }


# Interactive script for the user
if __name__ == "__main__":
    num1 = int(input("Enter first integer (a): "))
    num2 = int(input("Enter second integer (b): "))

    results = perform_all_operations(num1, num2)

    print("\n--- Results ---")
    for operation, result in results.items():
        print(f"{operation}: {result}")
```

```python
# After applying try-except
import math

def perform_all_operations(a: int, b: int) -> dict:
    """
    Performs addition, subtraction, multiplication, division, 
    and square root on two integer inputs.
    """
    # 1. Addition
    addition = a + b

    # 2. Subtraction
    subtraction = a - b

    # 3. Multiplication
    multiplication = a * b

    # 4. Division (safe check for zero)
    division = a / b if b != 0 else "Undefined (cannot divide by zero)"

    # 5. Square Roots (safe check for negative values)
    sqrt_a = math.isqrt(a) if a >= 0 and math.isqrt(a)**2 == a else (round(math.sqrt(a), 4) if a >= 0 else "Undefined (negative number)")
    sqrt_b = math.isqrt(b) if b >= 0 and math.isqrt(b)**2 == b else (round(math.sqrt(b), 4) if b >= 0 else "Undefined (negative number)")

    return {
        "Addition (a + b)": addition,
        "Subtraction (a - b)": subtraction,
        "Multiplication (a * b)": multiplication,
        "Division (a / b)": division,
        "Square Root of a": sqrt_a,
        "Square Root of b": sqrt_b,
    }


# Interactive script for the user
if __name__ == "__main__":
    try:
        num1 = int(input("Enter first integer (a): "))
        num2 = int(input("Enter second integer (b): "))

        results = perform_all_operations(num1, num2)

        print("\n--- Results ---")
        for operation, result in results.items():
            print(f"{operation}: {result}")

    except ValueError:
        print("Error: Please enter valid whole integers.")
```
