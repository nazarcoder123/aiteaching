### What are Python Keywords?
**Keywords** are special reserved words built directly into the core Python language syntax. They define the structural grammar and functional rules of Python programs.

Because they have strict, built-in meanings:
- You **cannot** use them as names for variables, functions, classes, or any custom identifiers.
- They are **case-sensitive** (e.g., `True`, `False`, `None` are capitalized; all others are lowercase).
- They cannot be reassigned or redefined.

---
## Getting the List of All Python Keywords

Python provides a built-in module named [`keyword`](https://docs.python.org/3/library/keyword.html) that exposes the complete list of reserved words via `keyword.kwlist`.

````python
# Inspecting all Python keywords using the keyword module
import keyword

print("Total number of keywords:", len(keyword.kwlist))
print("\nThe list of keywords are : ")
print(keyword.kwlist)
````

---
## How to Identify Python Keywords

1. **Syntax Highlighting in IDEs:**
   - Modern editors (VS Code, PyCharm, Jupyter) automatically apply distinct color styles (e.g. bold blue or purple) to reserved keywords.
2. **Programmatic Checking:**
   - Use the `keyword.iskeyword(string)` utility function.
3. **Observing `SyntaxError`:**
   - If you inadvertently use a keyword as an identifier, Python immediately raises a `SyntaxError` during tokenization/parsing.

````python
# Programmatically checking if a word is a keyword
test_words = ["for", "while", "student", "class", "score", "def", "total"]

for word in test_words:
    status = keyword.iskeyword(word)
    print(f"'{word:<10}' -> Is Keyword: {status}")
````

---
## Attempting to Use Keywords as Variable Names

If we attempt to use a keyword as a variable name, Python rejects it with a **SyntaxError**:

```python
for = 10
print(for)
```

**Output:**
```
File "<main.py>", line 1
    for = 10
        ^
SyntaxError: invalid syntax
```

Let's test this safely in Python using `exec()` inside a `try-except` block to inspect the error without crashing our notebook execution.

````python
# Demonstrating SyntaxError when attempting to use a keyword as a variable name
invalid_statements = [
    "for = 10",
    "class = 'Math'",
    "pass = True"
]

for stmt in invalid_statements:
    try:
        exec(stmt)
    except SyntaxError as err:
        print(f"Executing: '{stmt}'")
        print(f"❌ Caught SyntaxError: {err.msg} at line {err.lineno}\n")
````

````python
# Category 1: Value Keywords & Category 2: Operator Keywords
is_logged_in = True
has_permission = False
session_token = None

print("Value Keywords:")
print("is_logged_in   :", is_logged_in, "| Type:", type(is_logged_in))
print("session_token  :", session_token, "| Type:", type(session_token))

print("\nOperator Keywords in action:")
# 'and', 'or', 'not'
can_access = is_logged_in and not has_permission
print("is_logged_in and not has_permission:", can_access)

# 'is', 'in'
fruits = ["apple", "banana", "cherry"]
print("'apple' in fruits:", "apple" in fruits)
print("session_token is None:", session_token is None)
````

````python
# Category 3: Control Flow & Category 4: Exception Handling Keywords

marks = 85

# if, elif, else
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
else:
    grade = "C"

print(f"Marks: {marks} -> Grade: {grade}")

# try, except, finally
try:
    result = 100 / 0
except ZeroDivisionError as err:
    print("Caught division by zero:", err)
finally:
    print("Finally block executes cleanup unconditionally.")
````

````python
# Category 5: Function & Class, and Category 6: Context Management

# class and def keywords
class Greeter:
    def __init__(self, name):
        self.name = name
    
    # return keyword
    def greet(self):
        return f"Welcome to Python, {self.name}!"

# lambda keyword
square = lambda x: x ** 2

person = Greeter("Students")
print(person.greet())
print("Square of 8 (using lambda):", square(8))

# with, as keywords (Context Management)
with open("test_sample.txt", "w") as file:
    file.write("Python handles file closing automatically with 'with'!")

print("File written and closed safely.")

# Clean up temporary test file
import os
if os.path.exists("test_sample.txt"):
    os.remove("test_sample.txt")
````

## Keywords vs. Identifiers

Understanding the distinction between **Keywords** and **Identifiers** is a core programming milestone:

| Feature | Keywords | Identifiers |
| :--- | :--- | :--- |
| **Definition** | Reserved words built into the Python language with predefined meanings. | User-defined names given to variables, functions, classes, and modules. |
| **Usage as Variable Names** | **Cannot** be used as variable names or custom identifiers. | **Can** be used as variable names provided they adhere to identifier rules. |
| **Examples** | `if`, `else`, `for`, `while`, `def`, `class` | `x`, `number`, `sum`, `result`, `student_name` |
| **Role in Code** | Part of Python's grammar and syntax structure. | Meaningful labels chosen by the programmer to hold values or structures. |
| **Customizability** | Fixed by the language; cannot be redefined or changed. | Can be defined, reassigned, and modified freely by the programmer. |

*For detailed reference, see [Identifiers Guide](https://www.geeksforgeeks.org/python/python-keywords-and-identifiers/).*

---
## Variables vs. Keywords

| Feature | Variables | Keywords |
| :--- | :--- | :--- |
| **Purpose** | Used to store and reference data in memory. | Reserved words with language-level semantic meaning. |
| **Creation & Modification** | Can be created, updated, and deleted (`del`) by the programmer. | Cannot be created, modified, or deleted. |
| **Examples** | `x`, `age`, `user_name`, `total_price` | `if`, `while`, `for`, `return`, `import` |
| **Role** | Hold dynamic values that are manipulated during runtime. | Define the static syntax and control structure of the code. |
| **Naming Flexibility** | Flexible, subject only to naming rules (no digit prefix, etc.). | Strictly fixed and reserved by Python. |

````python
# Best Practices for Identifiers: Avoid Shadowing Built-ins

# 1. Identifiers should be descriptive
student_count = 30
average_score = 88.5

# 2. Never name identifiers similar to keywords or built-ins:
# Bad: list = [1, 2, 3]  (Shadows the built-in list type!)
# Good:
number_list = [1, 2, 3]

# Bad: id = 101          (Shadows the built-in id() function!)
# Good:
student_id = 101

print(f"Valid Identifiers -> ID: {student_id}, Count: {student_count}, Avg: {average_score}")
````

````python
# --- Student Practice Solutions ---
import keyword

# Solution 1 & 2: Programmatic Keyword Filter
sample_words = ["lambda", "my_lambda", "Global", "global", "pass", "password", "yield", "total_yield"]

keywords_found = []
valid_identifiers = []

for item in sample_words:
    if keyword.iskeyword(item):
        keywords_found.append(item)
    elif item.isidentifier():
        valid_identifiers.append(item)

print("Keywords found     :", keywords_found)
print("Valid identifiers  :", valid_identifiers)

print("-" * 50)

# Solution 3: Corrected Code
subject_def = "Mathematics"
class_level = 10
print("Corrected Identifiers:", subject_def, "| Class Level:", class_level)
````
