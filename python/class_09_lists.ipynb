### What is a Python List?
A **List** is a built-in sequence data structure used to store an ordered, mutable collection of items.

### Key Characteristics:
- **Mutable:** Elements can be updated, appended, inserted, or removed after creation.
- **Ordered:** Elements maintain the exact order in which they are added.
- **Heterogeneous:** A single list can store items of diverse data types (`int`, `str`, `float`, `bool`, or other lists).
- **Zero-Indexed:** First element is at index `0`.

---
## 1. Creating a List

Python provides multiple convenient ways to instantiate lists:

1. **Using Square Brackets `[...]`:** Direct declaration.
2. **Using the `list()` Constructor:** Converts any iterable (tuples, strings, sets) into a list.
3. **Using Repetition (`*`):** Creates pre-sized lists filled with default values.

```python
# 1. Creating lists using square brackets []
a = [1, 2, 3]
b = ["apple", "banana"]

print("List a (Integers):", a)
print("List b (Strings) :", b)

print("-" * 40)

# 2. Creating lists using the list() constructor
from_tuple = list((1, 2, 3, 'apple', 4.5))
from_string = list("GFG")

print("From Tuple :", from_tuple)
print("From String:", from_string)

print("-" * 40)

# 3. Creating lists with repeated elements using '*'
repeated_twos = [2] * 5
zeros_vector = [0] * 7

print("Repeated [2] * 5:", repeated_twos)
print("Repeated [0] * 7:", zeros_vector)
```

---
## 2. Internal Representation & Memory Model

A Python list is implemented internally as a **dynamic array of object pointers (references)**:
- The list does not store raw values directly contiguous in memory.
- Instead, each slot in the list holds the **memory address (`id`)** of the corresponding object.
- The actual objects (integers, strings, booleans) exist independently on the heap.
- This design allows Python lists to hold heterogeneous elements of varying sizes seamlessly.

```python
# Demonstrating Heterogeneous Elements & Reference Inspection
mixed_list = [1, 2, 2, "Python"]

print("mixed_list[0] :", mixed_list[0])
print("Entire list   :", mixed_list)

# Inspecting memory references of list elements
for idx, item in enumerate(mixed_list):
    print(f"Index {idx} -> Item: {repr(item):<10} | Type: {type(item).__name__:<6} | Object ID: {id(item)}")
```

## 3. Accessing Elements (Indexing & Slicing)

- **Zero-Based Positive Indexing:** `a[0]` (first element) to `a[len-1]` (last element).
- **Negative Indexing:** `a[-1]` (last element) to `a[-len]` (first element).
- **Slicing:** `a[start:end:step]` extracts a sublist.

```python
# Accessing elements via indexing and slicing
a = [10, 20, 30, 40, 50]

print("Original list:", a)
print("First element (a[0]) :", a[0])
print("Last element  (a[-1]):", a[-1])
print("Middle slice  (a[1:4]):", a[1:4])
print("Reversed list (a[::-1]):", a[::-1])
```

## 4. Adding Elements to a List

Python provides three primary methods to insert new items:

| Method | Syntax | Description |
| :--- | :--- | :--- |
| **`append(x)`** | `a.append(x)` | Adds a single element `x` to the **end** of the list. |
| **`insert(i, x)`** | `a.insert(i, x)` | Inserts element `x` at a specific index `i` (shifts existing items right). |
| **`extend(iterable)`** | `a.extend(iter)` | Unpacks and appends all items from an iterable to the end. |

> **Critical Distinction: `append()` vs. `extend()`**
> - `a.append([3, 4])` appends the list as a single nested element: `[1, 2, [3, 4]]`.
> - `a.extend([3, 4])` unpacks and adds individual items: `[1, 2, 3, 4]`.

```python
# Demonstrating append(), insert(), and extend()

# 1. append()
a = [1, 2]
a.append(3)
print("After append(3)       :", a)

# 2. insert()
a = [1, 3]
a.insert(1, 2)  # Insert value 2 at index 1
print("After insert(1, 2)    :", a)

# 3. extend()
a = [1, 2]
a.extend([3, 4])
print("After extend([3, 4])  :", a)

# Contrast: append vs extend
test_append = [1, 2]
test_append.append([3, 4])
print("Contrast - append([3, 4]):", test_append)
```

## 5. Updating Elements (List Mutability)

Because lists are **mutable**, elements can be reassigned in-place using their index.

```python
# Updating elements in-place
a = [10, 20, 30, 40, 50]
print("Before update:", a)

# Reassign index 1
a[1] = 25
print("After a[1] = 25:", a)

# Batch update via slicing
a[2:4] = [35, 45]
print("After slice update a[2:4] = [35, 45]:", a)
```

---
## 6. Removing Elements from a List

Python offers four complementary techniques to delete items:

1. **`remove(value)`:** Deletes the **first occurrence** of a specified value (raises `ValueError` if not found).
2. **`pop(index)`:** Removes and **returns** the element at index (defaults to the last element `-1`).
3. **`del a[index]`:** Built-in statement that deletes an item or slice at a specific index.
4. **`clear()`:** Empties the entire list, resulting in `[]`.

```python
# Demonstrating Methods to Remove Elements

# 1. remove(val): removes by value
a = [1, 2, 3, 2]
a.remove(2)  # Removes first 2
print("After remove(2) :", a)

# 2. pop(idx): removes and returns item
popped_item = a.pop()  # Removes last item
print(f"After pop()     : {a} (Popped value: {popped_item})")

# 3. del statement: removes by index
a = [10, 20, 30]
del a[1]  # Deletes element at index 1
print("After del a[1]  :", a)

# 4. clear(): removes all elements
a.clear()
print("After clear()   :", a)
```

## 7. Iterating Over Lists

Lists can be traversed:
1. Directly using a `for` loop.
2. With `enumerate()` to access both index and item.
3. Using **List Comprehensions** for concise transformations.

```python
# Iterating over lists
fruits = ['apple', 'banana', 'cherry']

print("Direct iteration:")
for item in fruits:
    print("-", item)

print("\nIndexed iteration with enumerate():")
for idx, item in enumerate(fruits):
    print(f"Item #{idx}: {item}")

print("\nList Comprehension (Transforming elements):")
uppercase_fruits = [item.upper() for item in fruits]
print("Uppercase fruits:", uppercase_fruits)
```

## 8. Nested Lists (2D Arrays & Matrices)

A **Nested List** is a list containing other lists as its elements. It is standard for representing matrices, grids, and tabular datasets.

### Accessing Nested Elements:
- `matrix[row_index]`: Selects the entire row list.
- `matrix[row_index][col_index]`: Selects the specific element inside the row.

```python
# Nested Lists: Matrices and 2D Grids
matrix = [
    [1, 2],
    [3, 4]
]

print("First row (matrix[0])     :", matrix[0])
print("Element at matrix[1][0]   :", matrix[1][0])

# Traversing 2D Matrix
print("\nPrinting 2x2 Matrix:")
for row in matrix:
    for cell in row:
        print(cell, end=" ")
    print()
```
