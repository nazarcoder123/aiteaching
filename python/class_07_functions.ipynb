### What are Functions?
**Functions** are self-contained, reusable blocks of code designed to perform a specific task.

### Core Benefits of Using Functions:
1. **Reusability:** Write logic once and execute it anywhere by calling the function.
2. **Modularity & Organization:** Breaks large programs into small, manageable, testable components.
3. **Maintainability:** Fix bugs or update behavior in a single place without duplicating code across files.

---
## 1. Defining and Calling a Function

In Python, a function is declared using the [`def` keyword](https://www.geeksforgeeks.org/python/python-def-keyword/) followed by the function name, parentheses `()`, and a colon `:`.

### Syntax:
```python
def function_name(parameters):
    # indented function body
    return value
```

To **call (invoke)** the function, write its name followed by parentheses containing any required arguments.

````python
# Defining a basic function
def fun():
    print("Welcome to GFG")

# Calling the function:
fun()

# In Python, functions are first-class citizens!
# You can assign a function to a variable and invoke it through that variable:
greet = fun
print("Calling through variable 'greet':")
greet()
````

## 2. Function Arguments and Parameters

- **Parameter:** The variable listed inside the function's definition header (placeholder).
- **Argument:** The actual value sent to the function when it is called.

If a function does not encounter an explicit `return` statement, it returns `None` by default.

````python
# Function with a parameter and a return statement
def evenOdd(x):
    if (x % 2 == 0):
        return "Even"
    else:
        return "Odd"

# Calling the function with different arguments
result1 = evenOdd(16)
result2 = evenOdd(7)

print("16 is:", result1)
print(" 7 is:", result2)
````

## 3. Types of Function Arguments in Python

Python provides versatile ways to pass arguments to functions:
1. **Default Arguments**
2. **Keyword Arguments (`key=value`)**
3. **Positional Arguments**
4. **Arbitrary Arguments (`*args` and `**kwargs`)**

### 3.1 Default Arguments
[Default arguments](https://www.geeksforgeeks.org/python/default-arguments-in-python/) provide fallback values if an argument is not supplied during the function call.
> **Rule:** Non-default parameters must always precede default parameters in the function definition header.

````python
# Demonstration of Default Arguments
def myFun(x, y=50):
    print("x: ", x)
    print("y: ", y)

# Calling with only one argument (uses default y=50)
print("--- Call with 1 argument ---")
myFun(10)

# Calling with both arguments (overrides default y)
print("\n--- Call with 2 arguments ---")
myFun(10, 99)
````

### 3.2 Keyword Arguments
Keyword arguments pass values explicitly using parameter names (`parameter=value`).
Because the parameter names are specified, **the order in which arguments are passed does not matter**.

````python
# Demonstration of Keyword Arguments
def student(fname, lname):
    print(f"First Name: {fname} | Last Name: {lname}")

# Order 1:
student(fname='Geeks', lname='Practice')

# Order 2 (Swapped parameter order produces identical result!):
student(lname='Practice', fname='Geeks')
````

### 3.3 Positional Arguments
Positional arguments are mapped to parameters based strictly on their **order (position)** in the function call.
If the order is changed, arguments will be assigned to unintended parameters.

````python
# Demonstration of Positional Arguments
def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)

print("Case-1 (Correct Order):")
nameAge("Olivia", 27)

print("\nCase-2 (Swapped Order causes incorrect association):")
nameAge(27, "Olivia")
````

### 3.4 Arbitrary Arguments: `*args` and `**kwargs`
When you don't know in advance how many arguments will be passed:
- **`*args` (Arbitrary Positional):** Collects excess positional arguments as a **tuple**.
- **`**kwargs` (Arbitrary Keyword):** Collects excess keyword arguments as a **dictionary**.

````python
# Demonstration of *args and **kwargs
def myFun(*args, **kwargs):
    print("Non-Keyword Arguments (*args) received as tuple:")
    for arg in args:
        print("  -", arg)

    print("\nKeyword Arguments (**kwargs) received as dictionary:")
    for key, value in kwargs.items():
        print(f"  {key} == {value}")

myFun('Hey', 'Welcome', first='Geeks', mid='for', last='Geeks')
````

## 4. Nested Functions (Function within Functions)

A function defined inside another function is called an [inner function](https://www.geeksforgeeks.org/python/python-inner-functions/) (or nested function).
- The inner function can read and access variables from the outer enclosing scope.
- Inner functions are commonly used for encapsulation and creating **decorators / closures**.

````python
# Demonstration of Inner / Nested Functions
def f1():
    s = 'I love GeeksforGeeks'
    
    def f2():
        print("Inside inner function f2():")
        print(s)  # Accesses variable 's' from enclosing outer scope
        
    f2()

f1()
````

## 5. The `return` Statement

The [`return` statement](https://www.geeksforgeeks.org/python/python-return-statement/) terminates function execution and transfers control (along with an optional value) back to the caller.

### Returning Multiple Values
In Python, a function can return multiple values separated by commas. Python automatically packs them into a **tuple**, which can be cleanly unpacked by the caller.

````python
# Demonstration of return values
def sq_value(num):
    return num ** 2

print("Square of 2 :", sq_value(2))
print("Square of -4:", sq_value(-4))

# Returning multiple values (Packed into tuple):
def calculate_stats(numbers):
    total = sum(numbers)
    count = len(numbers)
    avg = total / count
    return total, avg  # returns a tuple (total, avg)

data = [10, 20, 30, 40]
tot, average = calculate_stats(data)  # tuple unpacking
print(f"\nStats -> Total: {tot}, Average: {average}")
````

## 6. Pass by Object Reference (Mutability in Functions)

Python uses **Pass-by-Object-Reference** (also known as *Call-by-Sharing*):
- When you pass an argument, the function parameter receives a reference to the same object in memory.
- **Mutable Objects (Lists, Dictionaries):** Modifications made inside the function directly alter the original caller's object.
- **Immutable Objects (Integers, Strings, Tuples):** Reassigning the parameter creates a new local object; the original caller's value remains unaffected.

````python
# Case 1: Mutable object (List) modified inside function
def modify_list(x):
    x[0] = 20  # Mutates the caller's list in-place!

b = [10, 11, 12, 13]
print("List before function call:", b)
modify_list(b)
print("List after function call :", b)

print("-" * 45)

# Case 2: Immutable object (Integer) reassigned inside function
def modify_number(x):
    x = 20  # Rebinds local variable 'x' to a new integer object 20

a = 10
print("Integer before function call:", a)
modify_number(a)
print("Integer after function call :", a)
````

````python
# --- Student Practice Solutions ---

# Exercise 1: Default arguments
def greet_user(name, title="Student"):
    print(f"Hello, {title} {name}!")

greet_user("Alex")
greet_user("Sarah", title="Professor")

print("-" * 45)

# Exercise 2: *args adder
def sum_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

print("Sum of 3 numbers (10, 20, 30)        :", sum_numbers(10, 20, 30))
print("Sum of 5 numbers (1, 2, 3, 4, 5)     :", sum_numbers(1, 2, 3, 4, 5))

print("-" * 45)

# Exercise 3: Multi-value return
def convert_temp(celsius):
    fahrenheit = (celsius * 9/5) + 32
    kelvin = celsius + 273.15
    return fahrenheit, kelvin

c = 25
f, k = convert_temp(c)
print(f"{c}°C is equal to {f}°F and {k}K")
````
