### What are Data Types?
In Python programming, **Data Types** define the type of value stored in a variable and determine the specific operations that can be performed on that data.

Because Python treats **everything as an object**, every value is an instance of a specific built-in class:
1. **Efficiency:** Helps the Python interpreter understand how to allocate memory and process data efficiently.
2. **Safety & Correctness:** Ensures that only valid operations are performed on compatible types (e.g. preventing adding a string to an integer).

### Dynamic Typing Demonstration
In Python, you do not declare variable types explicitly. Reassigning a variable updates both its value and its underlying type:

```python
# Demonstrating dynamic typing across different data types
x = 50
print("Value:", x, "| Type:", type(x))

x = 60.5
print("Value:", x, "| Type:", type(x))

x = "Hello World"
print("Value:", repr(x), "| Type:", type(x))

x = ["geeks", "for", "geeks"]
print("Value:", x, "| Type:", type(x))

x = ("geeks", "for", "geeks")
print("Value:", x, "| Type:", type(x))
```

## 1. Numeric Data Types

[Python Numeric Data Types](https://www.geeksforgeeks.org/python/python-numbers/) are used to store numerical values. Python supports three primary numeric types:

1. **Integer (`int`):** Positive or negative whole numbers without decimals (e.g., `5`, `-42`). Python integers have arbitrary precision (they can grow as large as memory permits).
2. **Floating-Point (`float`):** Real numbers with a fractional component represented by a decimal point (e.g., `5.0`, `-0.003`, `3.14159`).
3. **Complex Number (`complex`):** Numbers with a real and an imaginary part written in the format `real + imag*j` (e.g., `2 + 4j`).

```python
# Numeric Data Types in Python
a = 5
b = 5.0
c = 2 + 4j

print("a =", a, "| Type:", type(a))
print("b =", b, "| Type:", type(b))
print("c =", c, "| Type:", type(c))

# Exploring complex numbers attributes
print("Real part of c:", c.real)
print("Imaginary part of c:", c.imag)
```

## 2. Sequence Data Types

A **Sequence** is an ordered collection of items (which can be of identical or mixed data types).
Because sequences are **ordered**, every element occupies a specific position called an **index**:
- **Positive Indexing:** Starts from `0` (left to right).
- **Negative Indexing:** Starts from `-1` (right to left).

The three primary sequence types in Python are:
1. Strings (`str`)
2. Lists (`list`)
3. Tuples (`tuple`)

### 2.1 String (`str`)
[Strings](https://www.geeksforgeeks.org/python/python-string/) store sequences of characters (text data). Strings are **immutable** (cannot be modified in-place) and can be created using single (`'...'`), double (`"..."`), or triple quotes (`'''...'''` / `"""..."""`).

```python
# String Data Type & Indexing
s = 'Welcome to the Geeks World'

print("String:", s)
print("Type:", type(s))

# Access string with positive and negative indexing:
print("Character at index 1 (s[1])  :", s[1])   # 'e'
print("Character at index -1 (s[-1]):", s[-1])  # 'd' (last character)
```

### 2.2 List (`list`)
[Lists](https://www.geeksforgeeks.org/python/python-lists/) are **ordered** and **mutable** collections.
- Defined using square brackets `[...]`.
- Elements can be of heterogeneous (different) data types.
- Can be modified, appended, or sorted in-place.

```python
# List Data Type & Indexing
a = [1, 2, 3]
print("List a:", a)

b = ["Geeks", "For", "Geeks", 4, 5]
print("List b:", b)

# Accessing elements via indexing
print("b[3]  (4th element) :", b[3])    # 4
print("b[-3] (3rd from end):", b[-3])   # 'Geeks'

# Demonstrating mutability (lists can be modified)
b[0] = "Python"
print("Modified list b:", b)
```

### 2.3 Tuple (`tuple`)
[Tuples](https://www.geeksforgeeks.org/python/python-tuples/) are **ordered** and **immutable** collections.
- Defined using parentheses `(...)`.
- Once created, elements **cannot** be added, removed, or reassigned.
- Useful for read-only data, constants, and dictionary keys.

> **Crucial Rule for Single-Element Tuples:**  
> A tuple with a single element **must include a trailing comma** (e.g. `(1,)`). Without the comma, Python treats `(1)` as an integer surrounded by parentheses!

```python
# Tuple Data Type & Single-Element Rule

# Single-element tuple vs integer
t_int = (1)      # Without trailing comma -> treated as int!
t1 = (1,)        # With trailing comma    -> properly treated as tuple!

print("Type of (1)  :", type(t_int))
print("Type of (1,) :", type(t1))

# Multi-element tuple
t2 = ('Geeks', 'For', 'Geeks', 1, 2)
print("\nTuple t2:", t2)
print("t2[3]  :", t2[3])
print("t2[-3] :", t2[-3])

# Tuples are immutable: attempting t2[0] = 'New' raises TypeError!
```

## 3. Boolean Data Type & Truthy/Falsy Values

The [Boolean](https://www.geeksforgeeks.org/python/boolean-data-type-in-python/) data type represents truth values: `True` or `False`. It is implemented as a subclass of integers (`bool`), where `True == 1` and `False == 0`.

### Truthy and Falsy Values
In Python, values in a conditional context evaluate to either truthy or falsy:
- **Falsy Values:** `False`, `None`, `0`, `0.0`, empty strings `""`, empty lists `[]`, empty tuples `()`, empty sets `set()`, empty dicts `{}`.
- **Truthy Values:** Any non-zero number, non-empty string, non-empty container.

```python
# Boolean Data Type & Truthy/Falsy Evaluation
print("Type of True :", type(True))
print("Type of False:", type(False))
print("-" * 35)

# Truthy / Falsy in conditional statements
if 1:
    print("1 is truthy")

if not 0:
    print("0 is falsy")

# Testing container emptiness
empty_list = []
if not empty_list:
    print("Empty list [] evaluates to falsy")

sample_text = "Python"
if sample_text:
    print("Non-empty string 'Python' evaluates to truthy")
```

## 4. Set Data Type (`set`)

[Sets](https://www.geeksforgeeks.org/python/sets-in-python/) are **unordered** and **mutable** collections of **unique** elements.
- Defined using curly braces `{...}` or the `set()` constructor.
- **No duplicates allowed:** Duplicate elements are automatically discarded.
- **Unordered:** Elements do not have a defined index position; therefore, `s[0]` is **invalid** and raises `TypeError`.
- Elements are typically accessed by iterating through them with a `for` loop or using membership operators (`in`).

```python
# Set Data Type: Automatic Deduplication & Iteration
s1 = {"a", "a", "b", "c", "b"}
print("Set s1 (duplicates automatically removed):", s1)

s2 = {"Geeks", "For", "Geeks"}
print("\nIterating through set s2:")
for item in s2:
    print("-", item)

# Testing membership
print("\nIs 'Geeks' in s2?", "Geeks" in s2)
```

## 5. Dictionary Data Type (`dict`)

[Dictionaries](https://www.geeksforgeeks.org/python/python-dictionary/) store data in **`key: value` pairs**.
- Defined using curly braces `{key: value}`.
- **Keys must be unique and immutable** (strings, numbers, tuples).
- **Values can be any data type** (including lists, sets, or nested dictionaries).
- Values are retrieved using either:
  1. Square bracket indexing: `d[key]` (raises `KeyError` if key is not found).
  2. The `.get(key, default)` method (returns `None` or a default fallback if key is missing).

> **Note on Case Sensitivity:** Dictionary keys are strictly case-sensitive (`'name'` and `'Name'` are treated as distinct keys).

```python
# Dictionary Data Type: Key-Value Access
d = {1: 'Geeks', 2: 'For', 3: 'Geeks'}

# Access using square brackets []
print("Access d[1]      :", d[1])

# Access using .get() method
print("Access d.get(2)  :", d.get(2))
print("Access d.get(99) :", d.get(99, "Key Not Found"))

# Demonstrating key case sensitivity
student = {
    "name": "Jordan",
    "Name": "Taylor"
}
print("\nCase-sensitive keys:")
print("student['name']:", student["name"])
print("student['Name']:", student["Name"])
```

## Summary Comparison of Python Collection Data Types

| Data Type | Syntax | Ordered? | Mutable? | Allows Duplicates? | Indexing Support |
| :--- | :---: | :---: | :---: | :---: | :---: |
| **String (`str`)** | `'hello'` | ✅ Yes | ❌ No | ✅ Yes | `s[0]`, `s[-1]` |
| **List (`list`)** | `[1, 2, 3]` | ✅ Yes | ✅ Yes | ✅ Yes | `lst[0]`, `lst[-1]` |
| **Tuple (`tuple`)** | `(1, 2, 3)` | ✅ Yes | ❌ No | ✅ Yes | `t[0]`, `t[-1]` |
| **Set (`set`)** | `{1, 2, 3}` | ❌ No | ✅ Yes | ❌ No (Unique only) | ❌ No indexing |
| **Dictionary (`dict`)** | `{'a': 1}` | ✅ Yes* | ✅ Yes | Keys: ❌ No, Values: ✅ Yes | `d['a']` (by Key) |

*\*Note: Starting in Python 3.7+, dictionaries preserve insertion order.*
