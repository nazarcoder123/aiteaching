### What is a Python Dictionary?
A **Dictionary** is an associative data structure that stores information in **`key: value` pairs**.

### Key Characteristics:
1. **Key Requirements:** Keys must be **unique** and **immutable (hashable)**, such as strings, numbers, or tuples.
2. **Value Flexibility:** Values can be of **any data type** (mutable or immutable, lists, sets, or nested dictionaries).
3. **Lookup by Key:** Access data by meaningful names/keys in $O(1)$ average time complexity, rather than numeric position.
4. **Insertion Order:** Since Python 3.7+, dictionaries preserve the insertion order of elements.

---
## 1. Creating a Dictionary

Dictionaries can be created:
1. Using curly braces `{key: value}`.
2. Using the built-in [`dict()`](https://www.geeksforgeeks.org/python/python-dict-function/) constructor with keyword arguments or lists of tuples.
3. Creating an empty dictionary using `{}` or `dict()`.

````python
# 1. Basic Dictionary using curly braces {}
data = {"name": "Jake", "age": 22}
print("Dictionary data:", data)

# 2. Multiple key-value pairs
a = {"x": 1, "y": 2}
print("Dictionary a   :", a)

# 3. Using the dict() constructor with keyword arguments
b = dict(name="Sam", age=20)
print("Dictionary b   :", b)

# 4. Using dict() with a list of (key, value) tuples
from_tuples = dict([('brand', 'Tesla'), ('model', 'Model 3')])
print("From Tuples    :", from_tuples)
````

---
## 2. Accessing Dictionary Items

Values are retrieved using their corresponding keys via:
1. **Square Brackets `d[key]`:** Fast, but raises a [`KeyError`](https://docs.python.org/3/library/exceptions.html#KeyError) if the key is not present.
2. **The [`.get(key, default)`](https://www.geeksforgeeks.org/python/python-dictionary-get-method/) Method:** Safe lookup that returns `None` (or an optional default value) when a key is missing.

````python
# Accessing Items: Square Brackets [] vs. .get()
d = {"name": "Kat", "age": 21}

# Accessing existing keys
print("d['name']    :", d["name"])
print("d.get('age') :", d.get("age"))

# Handling missing keys:
print("\nSafe lookup with .get() for missing key:")
print("d.get('gender')        :", d.get("gender"))
print("d.get('gender', 'N/A') :", d.get("gender", "Not Specified"))

# Attempting direct bracket access on missing key raises KeyError:
try:
    print(d["gender"])
except KeyError as err:
    print("\n[Caught Expected Error]: KeyError for missing key ->", err)
````

---
## 3. Adding and Updating Dictionary Items

- **Adding a New Key:** Assign a value to a non-existent key (`d[new_key] = value`).
- **Updating an Existing Key:** Assign a new value to an existing key (`d[existing_key] = new_value`).
- **Batch Updates:** The `.update({k: v, ...})` method adds or updates multiple key-value pairs simultaneously.

````python
# Adding and Updating Items
d = {"name": "Sam"}
print("Initial dictionary:", d)

# Adding a new key-value pair
d["age"] = 21
print("After adding 'age':", d)

# Updating an existing key's value
d["name"] = "Alex"
print("After updating 'name':", d)

# Batch update using .update()
d.update({"city": "New York", "role": "Engineer"})
print("After .update() batch update:", d)
````

---
## 4. Removing Dictionary Items

Python provides four specialized ways to remove items:

| Method | Syntax | Description |
| :--- | :--- | :--- |
| **`del`** | `del d[key]` | Deletes the item with the given key (raises `KeyError` if key not found). |
| **`pop()`** | `d.pop(key, default)` | Removes the key and **returns** its value. |
| **`popitem()`** | `d.popitem()` | Removes and returns the **last inserted** `(key, value)` tuple. |
| **`clear()`** | `d.clear()` | Empties the entire dictionary, leaving `{}`. |

````python
# Demonstrating Methods to Remove Dictionary Items

# 1. del statement
d1 = {"a": 1, "b": 2, "c": 3}
del d1["a"]
print("After del d1['a']:", d1)

# 2. pop(key): removes and returns value
d2 = {"a": 1, "b": 2}
popped_val = d2.pop("a")
print(f"After pop('a'): {d2} (Popped Value: {popped_val})")

# 3. popitem(): removes last inserted item
d3 = {"a": 1, "b": 2, "c": 3}
popped_item = d3.popitem()
print(f"After popitem(): {d3} (Removed Pair: {popped_item})")

# 4. clear(): removes all pairs
d4 = {"a": 1, "b": 2}
d4.clear()
print("After clear():", d4)
````

## 5. Iterating Through a Dictionary

A dictionary can be traversed with a `for` loop in three distinct ways:

1. **Iterating Keys:** Default loop iteration (`for key in d:` or `for key in d.keys():`).
2. **Iterating Values:** Using [`d.values()`](https://www.geeksforgeeks.org/python/python-dictionary-values/).
3. **Iterating Key-Value Pairs:** Using [`d.items()`](https://www.geeksforgeeks.org/python/python-dictionary-items-method/), which unpacks each pair as `(key, value)`.

````python
# Iterating Through a Dictionary
d = {"a": 1, "b": 2, "c": 3}

print("--- 1. Iterating Keys ---")
for key in d:
    print("Key:", key)

print("\n--- 2. Iterating Values ---")
for val in d.values():
    print("Value:", val)

print("\n--- 3. Iterating Key-Value Pairs (.items()) ---")
for key, value in d.items():
    print(f"{key} => {value}")
````

## 6. Nested Dictionaries

A [Nested Dictionary](https://www.geeksforgeeks.org/python/python-nested-dictionary/) is a dictionary containing one or more dictionaries as its values.

### Accessing Nested Data:
Chaining square brackets `dict[outer_key][inner_key]` or chained `.get()` methods allows deep navigation into hierarchical records.

````python
# Nested Dictionary Example: Student Record System
students = {
    "student1": {
        "name": "Sam",
        "age": 20,
        "grades": [88, 92, 95]
    },
    "student2": {
        "name": "Olivia",
        "age": 22,
        "grades": [90, 85, 94]
    }
}

# Accessing nested fields:
print("Student 1 Name:", students["student1"]["name"])
print("Student 1 Grades:", students["student1"]["grades"])
print("Student 1 Highest Grade:", max(students["student1"]["grades"]))

print("-" * 45)

# Adding a new record to nested structure:
students["student1"]["email"] = "sam@example.com"
print("Updated student1 record:", students["student1"])
````

## 7. Dictionary Comprehensions

Dictionary comprehensions provide a powerful, expressive way to transform or filter dictionaries.

### Syntax:
```python
{key_expr: value_expr for item in iterable if condition}
```

````python
# Dictionary Comprehensions

# 1. Squaring numbers
numbers = [1, 2, 3, 4, 5]
squares_dict = {x: x ** 2 for x in numbers}
print("Number squares dictionary:", squares_dict)

# 2. Filtering dictionary by value condition
scores = {"Alice": 95, "Bob": 68, "Charlie": 84, "David": 55}
passing_students = {name: score for name, score in scores.items() if score >= 70}
print("Passing students (score >= 70):", passing_students)
````

Create a dictionary `contacts = {"Alice": "123-456", "Bob": "789-101"}`.

Given `capitals = {"USA": "Washington", "France": "Paris", "Japan": "Tokyo"}`.

````python
# --- Student Practice Solutions ---

# Exercise 1: Contact Directory
contacts = {"Alice": "123-456", "Bob": "789-101"}
contacts["Charlie"] = "555-888"
contacts["Alice"] = "111-222"
diana_number = contacts.get("Diana", "Contact Not Found")
removed_bob = contacts.pop("Bob")

print("Updated contacts:", contacts)
print("Diana lookup    :", diana_number)
print("Removed Bob     :", removed_bob)

print("-" * 45)

# Exercise 2: Word Frequency Counter
sentence = "apple banana apple cherry apple banana"
word_counts = {}

for word in sentence.split():
    word_counts[word] = word_counts.get(word, 0) + 1

print("Word frequencies:", word_counts)

print("-" * 45)

# Exercise 3: Dictionary Inversion
capitals = {"USA": "Washington", "France": "Paris", "Japan": "Tokyo"}
inverted = {city: country for country, city in capitals.items()}
print("Original capitals:", capitals)
print("Inverted mapping :", inverted)
````
