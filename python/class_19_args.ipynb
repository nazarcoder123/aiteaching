# Args & Kwargs

This allow us to pass a variable number of arguments to a function 

```python
# *args: It allows a function to accept any number of positional arguments. The datatype is tuple
# without *args 

def add(a,b):
    return a+b 

print(add(1,1)) 
```

```python
# with *args 

def adds(*args):
    print(type(args))
    print(args)
    

print(adds(10,20,30,40,50))  
```

```python
# Adding Multiple numbers 

def addss(*args):
    total = 0
    for num in args:
        total = total + num 
    return total 

print(addss(10,20))
print(addss(10,20,30))
print(addss(10,20,30,40)) 
```

```python
# **kwargs: It allows a function to accept any number of keyword arguments. 
# without **kwargs 

def student(name,age):
    print(name) 
    print(age) 

student("Rahul",22) 
```

```python
# with **kwargs

def students(**kwargs):
    print(type(kwargs))
    print(kwargs) 

students(
    name="nazar",
    age=22,
    city="hyderabad",
    qualification="B.Tech"
)
```

```python
# using **kwargs diaplay student details

def students_details(**kwargs):
    for key, value in kwargs.items():
        print(key,":",value) 

students_details(
    name="nazar",
    age=25,
    city="hyderabad",
    qualification="B.Tech",
    company="wipro"
)
```

```python
#*args: Multiple positional arguments , dtype: Tuple, Argument format: 10,20,30 , Access: By index / loop
#**kwargs: Multiple keyword arguments , dtype: Dict, Access: By key / loop
```

```python
# Now lets Implement both *args & **kwargs 

def students_detail(*args,**kwargs): 
    print("Positional argument:", args) 
    print("keyword arguments:", kwargs) 

students_detail(
    "nazar",
    25,
    city="hyderbad",
    company="wipro"
)
```
