### What are Operators and Operands?
In Python programming, **Operators** are special symbols used to perform computations and operations on values and variables.

- **Operator:** The special symbol representing the operation (e.g., `+`, `-`, `*`, `/`).
- **Operand:** The data or variable on which the operator is applied (in `10 + 20`, `10` and `20` are operands).

### Categories of Python Operators:
1. Arithmetic Operators
2. Comparison (Relational) Operators
3. Logical Operators
4. Bitwise Operators
5. Assignment Operators
6. Identity Operators (`is`, `is not`)
7. Membership Operators (`in`, `not in`)
8. Ternary Operator (Conditional Expression)
9. Operator Precedence & Associativity

## 1. Arithmetic Operators

[Arithmetic operators](https://www.geeksforgeeks.org/python/python-arithmetic-operators/) are used to perform mathematical calculations.

| Operator | Name | Description | Example (`a=15, b=4`) |
| :---: | :--- | :--- | :--- |
| `+` | Addition | Adds two values | `a + b` $\rightarrow$ `19` |
| `-` | Subtraction | Subtracts right operand from left | `a - b` $\rightarrow$ `11` |
| `*` | Multiplication | Multiplies two values | `a * b` $\rightarrow$ `60` |
| `/` | Division | Floating-point division | `a / b` $\rightarrow$ `3.75` |
| `//` | Floor Division | Integer division (truncates decimal part) | `a // b` $\rightarrow$ `3` |
| `%` | Modulus | Returns the remainder of division | `a % b` $\rightarrow$ `3` |
| `**` | Exponentiation | Calculates power ($a^b$) | `a ** b` $\rightarrow$ `50625` |

> **Key Difference:** The regular division operator (`/`) always returns a **float** (e.g. `3.75`), whereas floor division (`//`) returns an **integer** (e.g. `3`).

````python
# Demonstration of Arithmetic Operators
a = 15
b = 4

print("Addition (+):", a + b)
print("Subtraction (-):", a - b)
print("Multiplication (*):", a * b)
print("Division (/):", a / b)
print("Floor Division (//):", a // b)
print("Modulus (%):", a % b)
print("Exponentiation (**):", a ** b)
````

## 2. Comparison (Relational) Operators

[Comparison operators](https://www.geeksforgeeks.org/python/python-object-comparison-is-vs/) compare two values and return a boolean result: either `True` or `False`.

| Operator | Meaning | Example (`a=13, b=33`) | Result |
| :---: | :--- | :--- | :---: |
| `>` | Greater than | `a > b` | `False` |
| `<` | Less than | `a < b` | `True` |
| `==` | Equal to | `a == b` | `False` |
| `!=` | Not equal to | `a != b` | `True` |
| `>=` | Greater than or equal to | `a >= b` | `False` |
| `<=` | Less than or equal to | `a <= b` | `True` |

````python
# Demonstration of Comparison (Relational) Operators
a = 13
b = 33

print("a > b  :", a > b)
print("a < b  :", a < b)
print("a == b :", a == b)
print("a != b :", a != b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)
````

## 3. Logical Operators

[Logical operators](https://www.geeksforgeeks.org/python/python-logical-operators/) are used to combine multiple conditional statements.

| Operator | Description | Truth Condition |
| :---: | :--- | :--- |
| `and` | Logical AND | Returns `True` only if **both** operands are True |
| `or` | Logical OR | Returns `True` if **at least one** operand is True |
| `not` | Logical NOT | Inverts the boolean value (`True` becomes `False`) |

### Precedence of Logical Operators:
1. `not` *(Highest precedence)*
2. `and`
3. `or` *(Lowest precedence)*

````python
# Demonstration of Logical Operators
a = True
b = False

print("a and b :", a and b)   # True and False -> False
print("a or b  :", a or b)    # True or False  -> True
print("not a   :", not a)     # not True       -> False
````

## 4. Bitwise Operators

[Bitwise operators](https://www.geeksforgeeks.org/python/python-bitwise-operators/) operate directly on the binary bits (`0` and `1`) of integers.

### Common Bitwise Operators in Python:
1. **Bitwise NOT (`~`)**: Inverts each bit. Formula: `~x = -(x + 1)`.
2. **Bitwise Shift (`>>`, `<<`)**: 
   - Right Shift (`>>`): Shifts bits to right (divides by powers of 2).
   - Left Shift (`<<`): Shifts bits to left (multiplies by powers of 2).
3. **Bitwise AND (`&`)**: Sets each bit to 1 if both corresponding bits are 1.
4. **Bitwise XOR (`^`)**: Sets each bit to 1 if only one of two corresponding bits is 1.
5. **Bitwise OR (`|`)**: Sets each bit to 1 if at least one of two bits is 1.

### Binary Breakdown for `a = 10` and `b = 4`:
- `a = 10` in binary: `0000 1010`
- `b = 4` in binary:  `0000 0100`
- `a & b` : `0000 0000` = `0`
- `a | b` : `0000 1110` = `14`
- `a ^ b` : `0000 1110` = `14`
- `~a`    : `-(10 + 1)` = `-11`
- `a >> 2`: `0000 0010` = `2`
- `a << 2`: `0010 1000` = `40`

````python
# Demonstration of Bitwise Operators
a = 10
b = 4

print("Binary of a (10):", bin(a))
print("Binary of b (4) :", bin(b))
print("-" * 35)

print("Bitwise AND (a & b) :", a & b)
print("Bitwise OR  (a | b) :", a | b)
print("Bitwise NOT (~a)    :", ~a)
print("Bitwise XOR (a ^ b) :", a ^ b)
print("Right Shift (a >> 2):", a >> 2)
print("Left Shift  (a << 2):", a << 2)
````

## 5. Assignment Operators

[Assignment operators](https://www.geeksforgeeks.org/python/assignment-operators-in-python/) assign the evaluated value from the right-hand side to the variable on the left-hand side. Python also supports compound assignment operators (e.g. `+=`, `-=`, `*=`).

| Operator | Example | Equivalent To |
| :---: | :---: | :--- |
| `=` | `b = a` | Simple assignment |
| `+=` | `b += a` | `b = b + a` |
| `-=` | `b -= a` | `b = b - a` |
| `*=` | `b *= a` | `b = b * a` |
| `/=` | `b /= a` | `b = b / a` |
| `<<=` | `b <<= a` | `b = b << a` |

````python
# Demonstration of Compound Assignment Operators
a = 10
b = a
print("Initial b = a       :", b)

b += a
print("After b += a (20)   :", b)

b -= a
print("After b -= a (10)   :", b)

b *= a
print("After b *= a (100)  :", b)

b <<= a
print("After b <<= a (102400):", b)
````

## 6. Identity Operators (`is` and `is not`)

[Identity operators](https://www.geeksforgeeks.org/python/python-membership-identity-operators-not-not/) test whether two variables point to the **same object in memory (same memory address)**.

- **`is`**: Returns `True` if both variables point to the identical object in memory.
- **`is not`**: Returns `True` if both variables point to different objects.

> **Crucial Difference: `==` vs `is`**
> - `==` tests for **equality of values** (do they have the same data?).
> - `is` tests for **identity** (do they share the same memory location via `id()`?).

````python
# Demonstration of Identity Operators
a = 10
b = 20
c = a

print("a is not b :", a is not b)  # True, because 10 and 20 are different objects
print("a is c     :", a is c)      # True, because c references the same object as a

# Checking memory IDs
print("Memory id(a):", id(a))
print("Memory id(b):", id(b))
print("Memory id(c):", id(c))
````

## 7. Membership Operators (`in` and `not in`)

[Membership operators](https://www.geeksforgeeks.org/python/python-membership-identity-operators-not-not/) check whether a given element exists within a sequence (such as a string, list, tuple, or set).

- **`in`**: Returns `True` if the value is present in the sequence.
- **`not in`**: Returns `True` if the value is **not** present in the sequence.

````python
# Demonstration of Membership Operators
x = 24
y = 20
my_list = [10, 20, 30, 40, 50]

# Testing 'not in'
if x not in my_list:
    print("x is NOT present in given list")
else:
    print("x is present in given list")

# Testing 'in'
if y in my_list:
    print("y is present in given list")
else:
    print("y is NOT present in given list")
````

## 8. Ternary Operator (Conditional Expressions)

The [Ternary operator](https://www.geeksforgeeks.org/python/ternary-operator-in-python/) allows you to evaluate conditions in a compact, single-line expression instead of writing a multiline `if-else` block. It was introduced in Python 2.5.

### Syntax:
```python
value_if_true if condition else value_if_false
```

````python
# Demonstration of Ternary Operator
a, b = 10, 20

# Single-line conditional assignment:
min_val = a if a < b else b

print("Smaller value is:", min_val)

# Another example: checking even or odd
number = 17
status = "Even" if number % 2 == 0 else "Odd"
print(f"The number {number} is {status}.")
````

## 9. Precedence and Associativity of Operators

[Operator precedence and associativity](https://www.geeksforgeeks.org/python/precedence-and-associativity-of-operators-in-python/) dictate the evaluation order when an expression contains multiple operators.

### A. Operator Precedence (Order of Operations)
Determines which operator executes first. For example, multiplication (`*`) has higher precedence than addition (`+`).
In logical expressions, `not` > `and` > `or`.

### B. Operator Associativity (Direction of Evaluation)
When two operators have the **same precedence**, associativity determines the direction in which they are evaluated:
- **Left-to-Right Associativity:** Most operators in Python (e.g., `+`, `-`, `*`, `/`).
- **Right-to-Left Associativity:** Exponentiation (`**`).
  - `2 ** 3 ** 2` evaluates as $2^{(3^2)} = 2^9 = 512$, **not** $(2^3)^2 = 8^2 = 64$.

````python
# Demonstration of Operator Precedence
expr = 10 + 20 * 30  # Multiplication occurs first (20*30=600), then addition (+10)
print("10 + 20 * 30 =", expr)

name = "Alex"
age = 0

# 'and' has higher precedence than 'or':
# Evaluated as: (name == 'Alex') or ((name == 'John') and (age >= 2))
# True or (False and False) -> True or False -> True
if name == "Alex" or name == "John" and age >= 2:
    print("Hello! Welcome.")
else:
    print("Good Bye!!")
````

````python
# Demonstration of Operator Associativity

# 1. Left-to-Right: Division and Multiplication have equal precedence
print("100 / 10 * 10 :", 100 / 10 * 10)  # (100 / 10) * 10 = 10.0 * 10 = 100.0

# 2. Left-to-Right: Subtraction and Addition
print("5 - 2 + 3      :", 5 - 2 + 3)       # (5 - 2) + 3 = 3 + 3 = 6

# 3. Using Parentheses to override default associativity
print("5 - (2 + 3)    :", 5 - (2 + 3))     # 5 - 5 = 0

# 4. Right-to-Left: Exponentiation (**)
print("2 ** 3 ** 2    :", 2 ** 3 ** 2)     # 2 ** (3 ** 2) = 2 ** 9 = 512
````
