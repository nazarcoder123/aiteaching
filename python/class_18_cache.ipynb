```python
import time

cache = {}

def get_data(student_id):

    if student_id in cache:
        print("Fetching data from cache...")
        return cache[student_id]

    print("Fetching data from database...")
    time.sleep(3)

    data = f"Student {student_id} details"

    cache[student_id] = data

    return data


print(get_data(101))
print(get_data(101))
print(get_data(101))
```

```python
from functools import lru_cache
import time

@lru_cache(maxsize=3)
def get_data(student_id):

    print("Fetching data...")

    time.sleep(2)

    return f"Student {student_id} details"


print(get_data(101))
print(get_data(101))
print(get_data(102))
print(get_data(101))
```

```python
@lru_cache(maxsize=3)
def calculate(number):
    return number * number
```

```python
from functools import lru_cache

@lru_cache(maxsize=3)
def square(number):
    print("Calculating...")
    return number * number


print(square(5))
print(square(5))
print(square(10))

print(square.cache_info()) 
```

# what is caching?

>  caching is a technique used to store the previously calculated or frequently accessed data temperarily 

>  so that we can also reuse it instead of performing the same operation again

```python
# without caching
import time 

def get_data():
    print("Fetching data from database...")
    time.sleep(5) 
    return "Student Data" 

print(get_data()) 
print(get_data()) 
print(get_data()) 
```

```python
# Now let's Implement caching with the help of dictionary 

import time 

cache = {} 

def get_data(student_id):

    if student_id in cache:
        print("Fetching from cache....")  
        return cache[student_id] 
    
    print("Fetching data from database....") 
    time.sleep(5) 

    data = f"student {student_id} details" 

    cache[student_id] = data 
    return data 

##-- with similar numbers--##
print(get_data(101)) # database 5 sec
print(get_data(101)) # Immediatly
print(get_data(101)) # Immediatly

##---with unique numbers---##
# print(get_data(101)) 
# print(get_data(102)) 
# print(get_data(103)) 
```

```python
# Function caching using functools.lru_cache: This is a built in library called functools to implement caching
```

```python
# @lru_cache: It is a decorator that automatically stores the results of function calls. 
# LRU: Least Recently used. 
# maxsize: It will store 3 cache only eg: 101 -> stored , 102 -> stored , 103 -> stored , here we are having maxsize = 3 so we are from 101 to 103. But we are having 104 
# so It is removeing the cache of 101-> adding a new cache of 104

from functools import lru_cache 
import time 

@lru_cache(maxsize=3)
def get_data(student_id):
    print("Fetching data....") 
    time.sleep(5) 
    return f"student {student_id} details"

# print(get_data(101)) # 2 sec
# print(get_data(101)) # Immediatly 
# print(get_data(102)) # 2 sec 
# print(get_data(101)) # Immediatly

print(get_data(101)) # 5 sec
print(get_data(102))  # 5 sec
print(get_data(103))   # 5 sec 
print(get_data(104))    # 5 sec
print(get_data(102))     # Imm
print(get_data(103))      # Imm

print(get_data(101)) # 5 sec
```
