```python
# what are generator in python ?
# A generator is a special type of function that produces values one at a time instead of returning all value at once 
# Generator uses a special keyword yield instead of return to produce values 
```

```python
def get_numbers():
    return [1,2,3,4,5] 

numbers = get_numbers() 
print(numbers) 
```

```python
# Now lets add the flavour of generator

def get_numbers():
    yield 1 
    yield 2
    yield 3
    yield 4
    yield 5

numbers = get_numbers()
print(next(numbers)) 
print(next(numbers)) 
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

```python
# for loop with yield 

def get_numbers():
    for i in range(1,60):
        yield i

for number in get_numbers(): 
    print(number) 
```

```python
# python comprehensions: A comprehension is a concise way to create a new collection from an existing iterable 
# list comprehension syntax is: [expression for item in iterable]
```

```python
# Normal Approach 

squares = [] 

for number in range(1, 6):
    squares.append(number ** 2) 
print(squares) 
```

```python
# using list comprehension
squares = [number ** 2 for number in range(1,6)]
print(squares)
```

```python
# Using list comprehension print even and odd numbers

numbers = [1,2,3,4,5,6,7,8]

even_num = [n for n in numbers if n%2 == 0]
print(even_num) 
```

```python
# Dictionary comprehension

sq = {n:n **2 for n in range(1,6)}
print(sq) 
```

```python
# set comprehension

sqrts = {n ** 2 for n in range(1,6)}
print(sqrts)
```

```python
# using list comprehension
squares = tuple(number ** 2 for number in range(1,6))
print(squares)
```

# what is decorator in Py?

- A decorator in python is a function that takes another function as input, adds extra behaviour to it and retur a new function

or

- A decorator is a higher-order function that wraps another function to extend its behaviour without changing its source.

```python
# @my_decorator replaces say_hello with wrapper, so calling say_hello() runs the extra prints around the original. 

def my_decorator(func):
    def wrapper():
        print("Before the function runs") 
        func() 
        print("After the function runs") 
    return wrapper 

# decorator @ 
@my_decorator
def say_hello():
    print("Hello") 

say_hello()
```
