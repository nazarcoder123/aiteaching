### What is a Python Tuple?
A **Tuple** is an ordered, immutable collection of elements.

### Key Characteristics:
1. **Immutable:** Once created, elements cannot be modified, replaced, added, or removed.
2. **Ordered:** Elements maintain a fixed positional order accessible via indexing.
3. **Heterogeneous:** Can hold items of diverse data types simultaneously (integers, strings, booleans, lists, dictionaries).
4. **Data Integrity:** Because they are immutable, tuples protect data from accidental modification and can be used as **keys in dictionaries**.

---
## 1. Creating a Tuple

Tuples are primarily created by enclosing items inside parentheses `(...)`, separated by commas, or by using the `tuple()` constructor.

> **Crucial Rule for Single-Element Tuples:**  
> A tuple with only one element **must have a trailing comma** (e.g. `('Geek',)`).  
> Without a comma, Python evaluates `('Geek')` as a plain string inside parentheses!

```python
# 1. Empty Tuple
empty_tup = ()
print("Empty Tuple:", empty_tup)

# 2. Tuple of Strings
tup_str = ('Geeks', 'For')
print("String Tuple:", tup_str)

# 3. Creating Tuple from a List
li = [1, 2, 4, 5, 6]
tup_from_list = tuple(li)
print("Tuple from List:", tup_from_list)

# 4. Creating Tuple from a String (splits into characters)
tup_from_str = tuple('Geeks')
print("Tuple from String:", tup_from_str)

# 5. Single-Element Tuple Demonstration
not_a_tuple = ("Geek")
is_a_tuple = ("Geek",)
print("\nType of ('Geek') :", type(not_a_tuple))
print("Type of ('Geek',):", type(is_a_tuple))
```

---
## 2. Creating Tuples with Mixed Data Types

Tuples can store arbitrary data types together, including nested lists and dictionaries.

```python
# Tuple containing heterogeneous data types
mixed_tup = (5, 'Welcome', 7.5, True, [1, 2, 3], {'key': 'value'})

print("Mixed Tuple:", mixed_tup)
print("Length of Mixed Tuple:", len(mixed_tup))

# Inspecting types of each element
for idx, item in enumerate(mixed_tup):
    print(f"Index {idx} -> {repr(item):<16} | Type: {type(item).__name__}")
```

---
## 3. Accessing Elements in a Tuple

Tuples support standard zero-based positive indexing (`0` to `n-1`), negative indexing (`-1` to `-n`), and slicing (`tuple[start:stop:step]`).

```python
# Accessing Tuple Elements via Indexing and Slicing
tup = tuple("Geeks")
print("Original Tuple:", tup)

# Accessing individual characters
print("First element (tup[0]) :", tup[0])
print("Last element  (tup[-1]):", tup[-1])

# Slicing
print("Slice tup[1:4]         :", tup[1:4])
print("Slice tup[:3]          :", tup[:3])
```

---
## 4. Tuple Unpacking

**Unpacking** extracts values from a tuple and binds them directly into individual variables in a single, clean statement.

> **Requirement:** The number of variables on the left-hand side must exactly match the number of elements in the tuple, unless the asterisk `*` operator is used.

```python
# Basic Tuple Unpacking
tup = ("Geeks", "For", "Geeks")

# Unpacking values into variables a, b, c
a, b, c = tup

print("Unpacked variable a:", a)
print("Unpacked variable b:", b)
print("Unpacked variable c:", c)
```

---
## 5. Concatenation of Tuples

Tuples can be combined using the `+` operator, returning a **brand new tuple**.

> **Note:** Only tuples can be concatenated with tuples. Attempting to concatenate a tuple with a list or integer raises a `TypeError`.

```python
# Demonstrating Tuple Concatenation
tup1 = (0, 1, 2, 3)
tup2 = ('Geeks', 'For', 'Geeks')

# Combining two tuples
tup3 = tup1 + tup2
print("Tuple 1:", tup1)
print("Tuple 2:", tup2)
print("Concatenated Tuple 3:", tup3)

# Type error check when concatenating different types:
try:
    invalid = tup1 + [4, 5]
except TypeError as err:
    print("\n[Caught Expected Error]:", err)
```

---
## 6. Slicing Tuples in Depth

Syntax: `tuple[start:stop:step]`
- `[:]`: Returns a copy containing all elements.
- `[start:]`: Slices from `start` to the end.
- `[4:9]`: Slices from index 4 up to (but excluding) index 9.
- `[::-1]`: Reverses the sequence of the tuple.

```python
# Advanced Slicing Operations
tup = tuple('GEEKSFORGEEKS')

print("Original Tuple   :", tup)
print("Slice tup[1:]    :", tup[1:])
print("Slice tup[4:9]   :", tup[4:9])
print("Reversed tup[::-1]:", tup[::-1])
```

## 7. Tuple Immutability & Deletion

Because tuples are immutable:
- You **cannot** delete individual elements (`del tup[0]` raises `TypeError`).
- You **cannot** reassign an element (`tup[0] = 'X'` raises `TypeError`).
- You can delete the **entire tuple variable** from namespace using [`del tup`](https://www.geeksforgeeks.org/python/python-del-to-delete-objects/). Subsequent access raises `NameError`.

```python
# Demonstrating Tuple Immutability and del Statement
tup = (0, 1, 2, 3, 4)

# 1. Attempting item modification
try:
    tup[0] = 99
except TypeError as err:
    print("Caught TypeError:", err)

# 2. Attempting item deletion
try:
    del tup[0]
except TypeError as err:
    print("Caught TypeError:", err)

# 3. Deleting entire tuple variable
del tup

# Attempting access after deletion:
try:
    print(tup)
except NameError as err:
    print("Caught NameError:", err)
    print("Explanation: The variable 'tup' has been deleted successfully.")
```

---
## 8. Extended Tuple Unpacking with Asterisk (`*`)

When unpacking a tuple of arbitrary length, the **asterisk operator (`*`)** collects multiple remaining elements into a **list**.

This allows you to extract specific leading/trailing elements while grouping the rest.

```python
# Extended Unpacking with '*' (Starred Expression)
tup = (1, 2, 3, 4, 5)

# a gets first item, c gets last item, b collects all middle items into a list:
a, *b, c = tup

print("First item (a)       :", a, f"| Type: {type(a).__name__}")
print("Middle items (*b)    :", b, f"| Type: {type(b).__name__}")
print("Last item (c)        :", c, f"| Type: {type(c).__name__}")

# Another example: Grab head and tail
first, *rest = (10, 20, 30, 40)
print(f"\nFirst: {first}, Rest: {rest}")
```

---
## 9. Advanced Concept: Mutable Objects Inside Tuples

While a tuple itself cannot have its references reassigned or elements added/removed, **if an element inside the tuple is a mutable object (like a list), that nested object can be modified in-place!**

```python
# Modifying a mutable list nested inside an immutable tuple
tup_with_list = (1, 2, ["apple", "banana"])
print("Before modifying nested list:", tup_with_list)

# We cannot replace the list object itself:
# tup_with_list[2] = ["new", "list"]  -> TypeError!

# But we CAN mutate the list that the tuple references:
tup_with_list[2].append("cherry")
tup_with_list[2][0] = "avocado"

print("After modifying nested list :", tup_with_list)
```
