### What is a String in Python?
A **String** is an ordered, immutable sequence of characters enclosed inside quotes. Characters can include letters, numbers, punctuation symbols, escape sequences, and spaces.

### Key Characteristics:
- **No Separate `char` Type:** Python treats an individual character simply as a string of length 1.
- **Ordered Sequence:** Every character can be accessed via an index.
- **Immutable:** Once created, individual characters cannot be changed in-place.

---
## 1. Creating Strings

Strings can be created using:
- Single quotes: `'...'`
- Double quotes: `"..."`
- Triple quotes (multi-line): `'''...'''` or `"""..."""` (preserves line breaks and tabs verbatim).

````python
# Creating Single-line Strings
a = 'GFG'
b = "GeeksForGeeks"

print("Single quote string:", a)
print("Double quote string:", b)

print("-" * 40)

# Creating Multi-line Strings using Triple Quotes
s_multi1 = """I am Learning
Python String on GeeksforGeeks"""
print(s_multi1)

print("-" * 40)

s_multi2 = '''I'm a
Geek'''
print(s_multi2)
````

---
## 2. Accessing Characters in a String (Indexing)

Because strings are ordered sequences, characters are accessed using square brackets `string[index]`.

### Two Indexing Directions:
1. **Positive Indexing (Left-to-Right):** Starts at `0` for the first character, up to `len(s) - 1`.
2. **Negative Indexing (Right-to-Left):** Starts at `-1` for the last character, down to `-len(s)`.

```
 String:  A   B   C   D   E   F
 Pos Idx: 0   1   2   3   4   5
 Neg Idx:-6  -5  -4  -3  -2  -1
```

> **Rules:**
> - Accessing an index outside boundaries raises an `IndexError`.
> - Indices must be integers; using floats or strings raises a `TypeError`.

````python
# Demonstrating Positive and Negative Indexing
s = "ABCDEF"

print("String:", s)
print("Total length:", len(s))
print("-" * 35)

# Positive indexing (from left)
print("s[0] (First character) :", s[0])
print("s[4] (5th character)   :", s[4])

# Negative indexing (from right)
print("s[-1] (Last character) :", s[-1])
print("s[-3] (3rd from end)   :", s[-3])
print("s[-5] (5th from end)   :", s[-5])

# Out of range test demonstration
try:
    print(s[10])
except IndexError as err:
    print("\n[Caught Expected Error]:", err)
````

## 3. String Slicing

[Slicing](https://www.geeksforgeeks.org/python/string-slicing-in-python/) extracts a substring from a string.

### Syntax:
```python
string[start:end:step]
```
- **`start`:** Starting index (inclusive). Defaults to `0`.
- **`end`:** Ending index (**exclusive** — character at `end` is NOT included). Defaults to length of string.
- **`step`:** Stride / increment value. Defaults to `1`. A negative step (like `-1`) traverses in reverse.

````python
# Demonstrating String Slicing
s = "ABCDEF"

print("Original string  :", s)
print("s[1:4] (Index 1 to 3) :", s[1:4])   # BCD
print("s[:3]  (Start to index 2) :", s[:3])   # ABC
print("s[3:]  (Index 3 to end)   :", s[3:])   # DEF
print("s[::2] (Every 2nd char)   :", s[::2])  # ACE

# Reversing a string using slicing step of -1
print("s[::-1] (Reversed string) :", s[::-1]) # FEDCBA
````

## 4. Looping Through Strings

Strings are iterable objects, allowing direct traversal character-by-character using a `for` loop.

````python
# Iterating through string characters
s = "ABCDEF"

print("Iterating character by character:")
for char in s:
    print(char, end=" ")
print()
````

## 5. String Immutability and Updating Strings

Strings in Python are **immutable**: their internal content cannot be modified once instantiated.

Attempting item assignment such as `s[0] = 'X'` raises a `TypeError`.

### How to "Update" Strings?
To modify a string, we construct a **new string** using:
1. Slicing and concatenation (`+`)
2. Methods like [`.replace()`](https://www.geeksforgeeks.org/python/python-string-replace/)

````python
# Demonstrating String Immutability
s = "aBCDEF"

# Attempting direct in-place mutation:
try:
    s[0] = "A"
except TypeError as err:
    print("Caught TypeError:", err)
    print("Strings cannot be modified in-place!\n")

# Method 1: Updating via slicing and concatenation
s_updated = "A" + s[1:]
print("Updated via concatenation:", s_updated)

# Method 2: Updating via .replace()
original = "ABCD EF"
replaced = original.replace("ABC", "abc")
print("Original string :", original)
print("Replaced string :", replaced)
````

## 6. Deleting a String

- Individual characters **cannot** be deleted (`del s[0]` raises `TypeError`).
- The entire variable name reference can be removed from memory using the [`del`](https://www.geeksforgeeks.org/python/python-del-to-delete-objects/) keyword.
- Accessing a variable after deletion raises a `NameError`.

````python
# Deleting Strings
s = "ABC"
print("Before deletion:", s)

# Deleting the variable reference
del s

# Accessing deleted variable
try:
    print(s)
except NameError as err:
    print("Caught NameError:", err)
    print("The variable 's' has been deleted from the current namespace.")
````

## 7. Essential Built-in String Methods

Python offers rich built-in methods for text manipulation:

| Method | Description | Example |
| :--- | :--- | :--- |
| `len(s)` | Returns total character count | `len("Geeks") -> 5` |
| `s.upper()` | Converts string to uppercase | `"abc".upper() -> "ABC"` |
| `s.lower()` | Converts string to lowercase | `"ABC".lower() -> "abc"` |
| `s.strip()` | Removes leading and trailing whitespace | `"  hi  ".strip() -> "hi"` |
| `s.replace(old, new)` | Replaces occurrences of a substring | `"cat".replace("c", "b") -> "bat"` |
| `s.split(sep)` | Splits string into a list of words | `"a,b".split(",") -> ['a', 'b']` |
| `s.startswith(prefix)` | Tests prefix | `"Hello".startswith("He") -> True` |

````python
# Demonstration of Common String Methods
s = "GeeksforGeeks"
print("Length of string (len)     :", len(s))

sample = "Hello World"
print("Uppercase (upper)          :", sample.upper())
print("Lowercase (lower)          :", sample.lower())

padded = "   ABC   "
print("Stripped whitespace (strip):", repr(padded.strip()))

text = "Python is fun"
print("Replaced text (replace)    :", text.replace("fun", "awesome"))

csv_row = "Apple,Banana,Orange,Mango"
print("Split into list (split)    :", csv_row.split(","))
````

## 8. Concatenation and Repetition Operators

- **Concatenation (`+`):** Combines two or more strings together.
- **Repetition (`*`):** Duplicates a string an integer number of times.

````python
# Concatenation and Repetition
s1 = "Hello"
s2 = "World"

# Concatenation with '+'
combined = s1 + " " + s2
print("Concatenation (+):", combined)

# Repetition with '*'
repeat_greeting = "Hello " * 3
print("Repetition (*)   :", repeat_greeting)
print("Separator line   :", "=" * 35)
````

---
## 9. String Formatting: f-strings & `.format()`

### 1. Formatted String Literals (`f-strings`) — Python 3.6+ (Recommended)
Prefix the string with `f` and place variables or expressions directly inside curly brackets `{}`.

### 2. The `.format()` Method
Uses `{}` placeholders inside the template string and injects values via arguments.

````python
# String Formatting Approaches

# Approach 1: Modern f-strings (Fastest & most readable)
name = "Jake"
age = 22
score = 95.8
print(f"Name: {name}, Age: {age}, Score: {score:.1f}")

# Arithmetic expressions inside f-strings:
print(f"Next year, {name} will be {age + 1} years old.")

print("-" * 45)

# Approach 2: str.format() method
template = "My name is {} and I am {} years old.".format("Emily", 22)
print(template)

# Positional and named placeholders with format():
info = "{title} {name} scored {points}%".format(title="Ms.", name="Alice", points=98)
print(info)
````

## 10. String Membership Testing (`in` and `not in`)

The `in` and `not in` operators check whether a substring exists within a string.
- Case-sensitive: `"geeks"` and `"Geeks"` are different.
- Returns boolean `True` or `False`.

````python
# Membership Testing in Strings
s = "GeeksforGeeks"

print("'Geeks' in s     :", "Geeks" in s)   # True
print("'GfG' in s       :", "GfG" in s)     # False
print("'Python' not in s:", "Python" not in s) # True
````

Write a function `is_palindrome(text)` that returns `True` if a word reads the same forward and backwards (ignoring case), e.g., `"Radar"`, `"madam"`.

### Exercise 2: Name Sanitizer
Given messy user input `raw_name = "   jOhN dOe   "`:

### Exercise 3: Vowel Counter
Write a function `count_vowels(sentence)` that counts the total number of vowels (`a, e, i, o, u`) in a given string.

````python
# --- Student Practice Solutions ---

# Exercise 1: Palindrome Checker
def is_palindrome(text):
    clean = text.lower()
    return clean == clean[::-1]

print("Is 'Radar' a palindrome? :", is_palindrome("Radar"))
print("Is 'Python' a palindrome?:", is_palindrome("Python"))

print("-" * 45)

# Exercise 2: Name Sanitizer
raw_name = "   jOhN dOe   "
clean_name = raw_name.strip().title()
print("Original :", repr(raw_name))
print("Sanitized:", repr(clean_name))

print("-" * 45)

# Exercise 3: Vowel Counter
def count_vowels(sentence):
    vowels = "aeiouAEIOU"
    count = 0
    for char in sentence:
        if char in vowels:
            count += 1
    return count

test_str = "GeeksforGeeks is awesome!"
print(f"Total vowels in '{test_str}':", count_vowels(test_str))
````
